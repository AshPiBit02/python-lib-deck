Processing a large image represented as a (1080,1920,3) array (height x width x RGB)

```python
import numpy as np 
np.random.seed(6)
image = np.random.randint(0, 256, size=(1080, 1920, 3), dtype=np.uint8)
```

```python
# Report the memory used by the image in MB.
print(f"Memory used: {image.nbytes/(1024**2):.2f} MB")
```

```python
# Convert to float32 for processing - report new memory usage.
new_image=image.astype(np.float32) 
print(f"Memory used: {new_image.nbytes/(1024**2):.2f} MB")
```

```python
# Noramlize pixel values to [0,1] in-place.
new_image/=255
print("Image shape",new_image.shape)
```

```python
# Convert to grayscale using the weighted formula: gray=0.29898*R + 0.5870*G + 0.1140*B - fully verctorize using the ... ellipsis or direct column indexing.
gray=0.2989*new_image[...,0] + 0.5870*new_image[...,1] + 0.1140*new_image[...,2]
print("Image Scaled")
```

```python
# Report the shape and dtype of the grayscale result.
print("Shape of grayscale: ",gray.shape)
print("Data type of grayscale: ",gray.dtype)
```

```python
# Compare memory of the original unit8 vs the grayscale float32.
print(f"Original Image memory: {image.nbytes/1024**2:.2f} MB")
print(f"Grayscale Image memory: {gray.nbytes/1024**2:.2f} MB")
```

Although grayscale reduces the number of channels (from 3 to 1), the conversion to float32 increases the per‑pixel storage from 1 byte (uint8) to 4 bytes (float32).
That’s why the grayscale image ends up using more memory than the original RGB image.
