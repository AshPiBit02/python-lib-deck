Processing a large image represented as a (1080,1920,3) array (height x width x RGB)

```python
import numpy as np 
np.random.seed(6)
image = np.random.randint(0, 256, size=(1080, 1920, 3), dtype=np.uint8)
```

```python
# Report the memory used by the image in MB.
print(f"Memory used: {image.nbytes/(1024**2):.2f} MB")
```

```python
# Convert to float32 for processing - report new memory usage.
new_image=image.astype(np.float32) 
print(f"Memory used: {new_image.nbytes/(1024**2):.2f} MB")
```
