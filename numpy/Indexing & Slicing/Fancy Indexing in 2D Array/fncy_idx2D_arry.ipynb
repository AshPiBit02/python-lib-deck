# Fancy Indexing on 2D Arrays

```python
import numpy as np 
arr=np.arange(1,10,1).reshape(3,3)
print(arr)
```

```python
# Extract elements [1,5,9]
print(arr[[0,1,2],[0,1,2]])
```

```python
# Extract elements [3,5,7]
print(arr[[0,1,2],[2,1,0]])
```

```python
# Extract elements[2,4,8]
print(arr[[0,1,2],[1,0,1]])
```

```python
# Extract the first and third rows.
print(arr[[0,2]])
```

```python
# Extract the first and third columns.
print(arr[:,[0,2]])
```

```python
# Create a new array containing [1,6,8] using row and column index arrays.
new_arr=arr[[0,1,2],[0,2,1]]
print(new_arr)
```
