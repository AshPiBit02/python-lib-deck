# Indexing 2D Arrays

```python
import numpy as np 
rows=4
cols=4
matrix=np.array([[10*(i+1) + j+1 for j in range(cols)]for i in range(rows)])
print(matrix)
```

```python
# Access element 22
print(matrix[1][1])
```

```python
# Access element 34.
print(matrix[2][3]) # or matrix[2,3]
```

```python
# Access element 41. 
print(matrix[3][0])
```

```python
# Access the last elemement.
print(matrix[-1][-1])
```

```python
# Access the entire second row.
second_row=matrix[1] 
print(matrix[1:2,:]) # 2D vector
print(second_row) # 1D array
```

```python
# Access the entire third column.
third_col=matrix[:,2]
print(third_col)
```

```python
# Find the sum of the diagonal elements.
dia_sum=np.trace(matrix)
print(dia_sum)
```

```python
# Replace 33 with 100
matrix[2,2]=100
print(matrix)
```

```python
# Access element 24 using negative indexing.
print(matrix[-3,-1])
```
