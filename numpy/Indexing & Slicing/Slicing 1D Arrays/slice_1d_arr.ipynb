# Slicing 1D Arrays

```python
import numpy as np 
arr=np.arange(5,51,5)
print(arr)
```

```python
# Exract the first five elements.
print(arr[:5])
```

```python
# Extract the last four elements.
print(arr[-4:])
```

```python
# Extract every second element.
print(arr[1::2])
```

```python
# Reverse the array using slicing.
print(arr[::-1])
```

```python
# Extract only odd-indexed elements.
print(arr[1::2])
```

```python
# Extract only even-indexed elements.
print(arr[::2])
```

```python
# Extract [20,25,30,35]
print(arr[3:7])
```
