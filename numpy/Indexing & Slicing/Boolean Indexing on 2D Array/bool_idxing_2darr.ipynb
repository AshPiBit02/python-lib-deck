# Boolean Indexing on 2D Arrays

```python
import numpy as np 
arr=np.arange(10,91,10).reshape(3,3)
print(arr)
```

```python
# Extract all values greater than 50.
print(arr[arr>50])
```

```python
# Extract all values less than 40.
print(arr[arr<40])
```

```python
# Count values greater than 30.
print((arr>30).sum())
```

```python
# Extract all multiples of 20.
print(arr[arr%20==0])
```

```python
# Replace values greater than 60 with -1.
arr[arr>60]=-1
print(arr)
```

```python
# Extract all even numbers.
print(arr[arr%2==0])
```
