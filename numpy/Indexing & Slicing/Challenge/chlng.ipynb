```python
import numpy as np 
```

```python
# Create a 6x6 matrix and extract its center 2x2 block.
org_matrix=np.arange(1,37,1).reshape(6,6)
print(org_matrix)
```

```python
sub_matrix=org_matrix[2:4,2:4]
print(sub_matrix)
```

```python
# Reverse only the rows of a matrix.
rev_row_mat=org_matrix[::-1,:]
print(rev_row_mat)
```

```python
# Extract all elements greater than the mean of the array.
print(org_matrix[org_matrix.mean()<org_matrix])
```

```python
# Using fancy indexing, swap the first and last rows of a matrix.
swap_matrix=org_matrix[[5,1,2,3,4,0],:]
print(swap_matrix)
```

```python
# Extract the border elements of a 5x5.
new_matrix=org_matrix[:5,:5]
print(new_matrix)
```

```python
top=new_matrix[0,:]
print(top)
```

```python
botton=new_matrix[len(new_matrix)-1,:]
print(botton)
```

```python
left_right=new_matrix[1:len(new_matrix)-1,[0,len(new_matrix)-1]]
print(left_right)
```

```python
result=np.concat(top,botton,left_right.flatten())
print(result)
```
