# Slicing 2D Arrays

```python
import numpy as np
arr=np.arange(1,26).reshape(5,5)
print(arr)
```

```python
# Extract the first row.
print(arr[0,:])
```

```python
# Extract the last row.
print(arr[-1,:])
```

```python
# Extract the first column.
print(arr[:,0])
```

```python
# Extract the last column.
print(arr[:,-1])
```

```python
# Extract the top-left 3x3 submatrix.
submatrix=arr[:3,:3]
print(submatrix)
```

```python
# Extract the bottom-right 2x2 submatrix
print(arr[-2:,-2:])
```

```python
# Extract the rows 1-3 and columns 2-4.
print("Rows: \n",arr[:3,1:4])
```

```python
# Extract all even-numbered rows.
print(arr[::2,:])
```

```python
# Extract all odd-numbered colunns.
print(arr[:,1::2])
```

```python
# Reverse both rows and columns
print(arr[::-1,::-1])
```
