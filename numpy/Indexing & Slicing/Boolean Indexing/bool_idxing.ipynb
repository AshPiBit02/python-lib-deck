# Boolean Indexing

```python
import numpy as np
arr=np.array([12,5,18,7,25,30,9,14])
```

```python
# Extract elements greater than 15.
print(arr[arr>15])
```

```python
# Extract elements less than 10.
print(arr[arr<10])
```

```python
# Extract odd numbers.
print(arr[arr%2!=0])
```

```python
# Extract even numbers.
print(arr[arr%2==0])
```

```python
# Extract elements between 10 and 25
print(arr[(arr>10) & (arr<25)])
```

```python
# Count how many elements are greater than 15.
print((arr>15).sum())
```

```python
# Replace all elements less than 10 with 0.
arr[arr<10]=0
print(arr)
```

```python
# Find all elements divisible by 3.
print(arr[arr%3==0])
```
