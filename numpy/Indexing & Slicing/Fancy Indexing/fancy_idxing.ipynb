# Fancy Indexing

```python
import numpy as np 
arr=np.arange(100,801,100)
print(arr)
```

```python
# Exract[100,300,500].
print(arr[[0,2,4]])
```

```python
# Extract [800,600,400].
print(arr[[7,5,3]])
```

```python
# Extract first,middle,and last elements.
print(arr[[0,(len(arr)//2),-1]])  # // integer division operator
```

```python
# Create a new array containing elements at indices[1,3,5,7]
new_arr=arr[[1,3,5,7]]
print(new_arr)
```

```python
# Repeat the first element three times using fancy indexing.
print(arr[[0,0,0]])
```
