# Indexing 1D Arrays

```python
import numpy as np
arr=np.arange(10,81,10)
print(arr)
```

```python
# Access the first element
print(arr[0])
```

```python
# Access the last element using negative indexing.
print(arr[-1])
```

```python
# Access the 4th element.
print(arr[3])
```

```python
# Find the sum of the first and last elements.
print("Sum : ",arr[0]+arr[-1])
```

```python
# Replace the 40 with 100.
arr[3]=100
print(arr)
```

```python
# Print all elements one by one using indexing.
for i in range(len(arr)):
    print(arr[i])
```
