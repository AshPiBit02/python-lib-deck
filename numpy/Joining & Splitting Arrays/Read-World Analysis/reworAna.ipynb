# Working with a sensor network that logs readings every hour. Reveive data in two separate shifts(morining and evening), each with 3 sensors.

```python
import numpy as np 
morning = np.array([
    [22.1, 45.3, 0.91],
    [22.5, 46.1, 0.89],
    [23.0, 47.5, 0.87],
    [23.8, 48.2, 0.85],
    [24.1, 49.0, 0.84],
    [24.5, 49.8, 0.82]
])  # 6 hours × 3 sensors

evening = np.array([
    [26.3, 55.1, 0.78],
    [25.9, 54.3, 0.80],
    [25.4, 53.0, 0.81],
    [24.8, 51.7, 0.83],
    [24.2, 50.5, 0.84],
    [23.7, 49.2, 0.86]
])  # 6 hours × 3 sensors
```

```python
# Combine both shifts into a full 12-hours log.
fullShift=np.vstack([morning,evening])
print("Full Shift(12-hour) Sensor Readings:\n",fullShift)
```

```python
# Split the full log into 3 equal time blocks.
firstBlock,secondBlock,thirdBlock=np.vsplit(fullShift,3)
print("First time block log: \n",firstBlock)
print("Second time block log: \n",secondBlock)
print("Third time block log: \n",thirdBlock)
```

```python
# For each block, compute the mean of each sensor.
Block1Mean=firstBlock.mean(axis=0)
Block2Mean=secondBlock.mean(axis=0)
Block3Mean=thirdBlock.mean(axis=0)

print("First Blocks' sensors mean: ",Block1Mean)
print("Second Blocks' sensors mean: ",Block2Mean)
print("Third Blocks' sensors mean: ",Block3Mean)
```

```python
# Stack the 3 mean rows into a summary array of shape(3,3) -> each row=one block's averages
summary_arr=np.vstack([Block1Mean,Block2Mean,Block3Mean])
print("The mean summary array : \n",summary_arr)
```

# Real World Problem
   # Working at a retail company tracking weekly sales across 5 products      categories   over 4 weeeks. Data arrives in two parts: first 2 weeks and last 2 weeks.

```python
weeks_1_2 = np.array([
    [1200, 850,  300, 1750, 640],
    [1350, 920,  275, 1800, 710]
])

weeks_3_4 = np.array([
    [980,  870,  410, 1600, 780],
    [1100, 950,  390, 1900, 820]
])
category=["Electornics","Clothing","Stationary","Food","Sports"]
```

```python
# Join both parts into a full monthly dataset.
fullMonth=np.vstack([weeks_1_2,weeks_3_4])
print("Overall Month Dataset: \n",fullMonth)
```

```python
# Split the dataset into individual weeks.
week1,week2,week3,week4=np.vsplit(fullMonth,4)
print("First week Datasets: ",week1)
print("Second week Datasets: ",week2)
print("Third week Datasets: ",week3)
print("Last week Datasets: ",week4)
```

```python
# For each week, find the index of the best-selling and worst-selling category.
print("Week1: ")
print('-'*55)
print(f"Best selling category is {category[np.argmax(week1)]} with sale of {week1.max()}$")
print(f"Worst selling category is {category[np.argmin(week1)]} with sale of {week1.min()}$\n")

print("Week2:")
print('-'*55)
print(f"Best selling category of is {category[np.argmax(week2)]} with sale of {week2.max()}$")
print(f"Worst selling category of is {category[np.argmin(week2)]} with sale of {week2.min()}$\n")

print("Week3:")
print('-'*55)
print(f"Best selling category of is {category[np.argmax(week3)]} with sale of {week3.max()}$")
print(f"Worst selling category of is {category[np.argmin(week3)]} with sale of {week3.min()}$\n")

print("Week4:")
print('-'*55)
print(f"Best selling category of is {category[np.argmax(week4)]} with sale of {week4.max()}$")
print(f"Worst selling category of is {category[np.argmin(week4)]} with sale of {week4.min()}$\n")
```

```python
# Identify which categories has sales below the overall monthly average across all weeks.
overall_avg=fullMonth.mean()
avg_per_category=fullMonth.mean(axis=0)
category_below_avg=np.where(avg_per_category<overall_avg,True,False)
print(category_below_avg)
# print("Categories' sales below the overall monthly average across all weeks: \n",category*category_below_avg)
print(overall_avg)
print(avg_per_category)
```
