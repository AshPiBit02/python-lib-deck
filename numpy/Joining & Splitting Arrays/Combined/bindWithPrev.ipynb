# Given a dataset of 8 student exam records(name_id,math,dsa,statistics)

```python
import numpy as np 
students = np.array([
    [101, 88, 76, 91],
    [102, 72, 85, 68],
    [103, 95, 90, 87],
    [104, 60, 70, 75],
    [105, 83, 78, 88],
    [106, 91, 95, 92],
    [107, 55, 62, 70],
    [108, 78, 80, 85]
])
print("Students Data: \n",students)
```

```python
# Split the data into two batches of 4 students each.
batch1,batch2=np.vsplit(students,2)
print("First Batch Student Info: \n",batch1)
print("Second Batch Students Info: \n",batch2)
```

```python
# From batch1, select only students who scored above 80 in math.
filtered_batch1=batch1[batch1[:,1]>80]
print("First batch students with marks above 80 in math are: ",filtered_batch1[:,0])
```

```python
# From batch2, select only students who scored above 80 in statistics.
filtered_batch2=batch2[batch2[:,3]>80]
print("Second batch students with marks above 80 in statistics are: ",filtered_batch2[:,0])
```

```python
# Join both filtered results back into one array.
filtered_students=np.vstack([filtered_batch1,filtered_batch2])
print("Filtered Students' info: \n",filtered_students)
```
