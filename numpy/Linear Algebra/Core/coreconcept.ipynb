# Two vectors representing a user's feature profile and item weights in a recommendation system:

```python
import numpy as np 
user   = np.array([0.8, 0.3, 0.9, 0.5, 0.7])
item   = np.array([0.6, 0.9, 0.4, 0.8, 0.2])
```

```python
# Compute the dot product.
raw_dot=np.dot(user,item)
print("The Dot product of user and item is: ",np.round(raw_dot,2))
```

```python
# Normalize both vectors(divide each by its own magnitude using np.linalg.norm())
user_norm=user/np.linalg.norm(user)
item_norm=item/np.linalg.norm(user)
print("Normalized user vector: ",user_norm)
print("Normalized item vector: ",item_norm)
```

```python
# Dot product of normalized vectors
cosine_sim=np.dot(user_norm,item_norm)
print("Cosine similarity: ",cosine_sim)
```

What does the second result represents and why is it bounded between -1 and 1?
 * The second result represents normalized dot product which is called cosine similirity.
 * It is bounded between -1 and 1 because it is mathematically defined as the cosine of the angle between two vectors. 

# A transformation matrix and a batch of 4 data points(each row is a point)

```python
T = np.array([[2, 0],
              [0, 3]])

points = np.array([[1, 1],
                   [2, 1],
                   [1, 3],
                   [2, 2]])
```

```python
# Apply the transformation to all points in one operation.
result=points @ T
print("Original points: \n",points)
print("The Transformed points: \n",result)
print("\n Result shape: ",result.shape)
```

# Three matrics:

```python
A = np.array([[4, 7], [2, 6]])
B = np.array([[1, 0], [0, 1]])
C = np.array([[2, 4], [1, 2]])
```

```python
# Compute determinant for each.
detA=np.linalg.det(A)
detB=np.linalg.det(B)
detC=np.linalg.det(C)
print("Determinant of A: ",np.round(detA,2))
print("Determinant of B: ",detB)
print("Determinant of C: ",detC)
```

Matrix C is invertible because it is a singular matrix(i.e. det=0)

```python
# Invert possible matrices
invA=np.linalg.inv(A)
invB=np.linalg.inv(B)
print("Inverse of A : \n",invA)
print("Inverse of B : \n",invB)
```
