# A dataset of 5 houses with 3 features each(size,rooms,age):

```python
import numpy as np
X = np.array([
    [1500, 3, 10],
    [2000, 4,  5],
    [1200, 2, 20],
    [1800, 3,  8],
    [2500, 5,  3]
], dtype=float)
```

```python
# Analysis
max_size=X[:,0].max()
min_size=X[:,0].min()

max_rooms=X[:,1].max()
min_rooms=X[:,1].min()

max_age=X[:,2].max()
min_age=X[:,2].min()

print(f"Max House size: {max_size} sq. ft")
print(f"Min House size: {min_size} sq. ft")

print(f"Max Rooms: {max_rooms}")
print(f"Min Rooms: {min_rooms}")

print(f"Max House Age: {max_age} year(s)")
print(f"Min House Age: {min_age} year(s)")
```

```python
# Normalize each feature column to range [0,1] using: (x-min)/(max-min)
X_norm=np.round((X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0)),4)
print(X_norm)
```

```python
# Compute the covariance matrix as (XnT & Xn)/(n-1) wher Xn is the normalized data.
XnT=np.transpose(X_norm)
X_cov=(np.dot(XnT,X_norm))/(len(X_norm)-1)
print("The Covariance matrix: \n",X_cov)
```

```python
# Verify teh covariance matrix is symmetric by checking np.allclose(cov,cov.T)
X_covT=np.transpose(X_cov)
print("Is covarience matrix symmetric?: ",np.allclose(X_cov,X_covT))
```

# A weight matrix for a neural network layer and an input batch:

```python
np.random.seed(7)
W = np.random.randn(4, 3)   # 4 neurons, 3 input features
X = np.random.randn(5, 3)   # 5 samples, 3 features
b = np.array([0.5, -0.5, 0.3, 0.1])  # bias for each neuron
```

```python
# Compute the forward pass: X @ W_T + b
W_T=np.transpose(W)
output=X @ W_T +b
print("Output: \n",output)
```

```python
# From the output, select only the rows(samples) where all newron values are positive.
selected_samples = output[(output > 0).all(axis=1)]
print(selected_samples)
```

```python
# Report how many samples passed that condition.
print(f"{selected_samples.shape[0]} samples that passed the previous condition. ")
```

# A 4x4 matrix of quiz scores between students(entry[i,j]=score student i gave to student j):

```python
scores = np.array([
    [0,  85, 90, 70],
    [80,  0, 75, 95],
    [88, 72,  0, 85],
    [91, 68, 82,  0]
])
```

```python
# Extract the upper traingle(excluding diagonal) - these are socres given by lower-numbered students and vice versa.
upr_matrix=np.triu(scores,k=1)
lwr_matrix=np.tril(scores,k=-1)
print("Scores of upper numbered students given by lowered ones: \n",upr_matrix)
print("Scores of lower numbered students given by upper ones: \n",lwr_matrix)
```

```python
# Compute the mean of each group and find the which group gave higher scores on average.
upr_mean=upr_matrix.mean()
lwr_mean=lwr_matrix.mean()
def result(mean,group):
    print(f"{group} Group gave higher socres on average(i.e. {mean})")
if upr_mean>lwr_mean :
    result(upr_mean,"Lower")
else:
    result(lwr_mean,"Upper")
```

# A system of 3 linear equations.
* 2x + y - z  = 8,
* 3x - y + 2z = -11,
* -2x + y + 2z = -3

```python
A = np.array([[ 2,  1, -1],
              [-3, -1,  2],
              [-2,  1,  2]], dtype=float)

b = np.array([8., -11., -3.])
```

```python
# Verify the systems has a unique solution by checking det(A).
det_A=np.linalg.det(A)
print("Determinant of A: ",det_A) 
if det_A!=0:
    print("Since the determinant of A doesn't equalt to zero so, the system has unique solution.")
```

```python
# Solve using np.linalg.solve()
x=np.linalg.solve(A,b)
print("Value of x, y and z are: ",x)
```

```python
# Verify the solution by computing A @ x and comparing to b using np.allclose().
cross=A @ x
print("Is A @ x and b equal?: ",np.allclose(cross,b))
```
