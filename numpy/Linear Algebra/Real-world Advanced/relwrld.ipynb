# A basic Principle Component Analysis from scratch on a small dataset:

```python
import numpy as np 
np.random.seed(0)
X = np.random.randn(100, 4)   # 100 samples, 4 features
print("Dataset: \n",X)
```

```python
# Compute the data by subtracting the column means.
col_mean=X.mean(axis=0)
print("Column means: ",col_mean)
X_centered=X-col_mean
print("Centered data: \n",X_centered)
```

```python
# Compute the covariance matrix: C = (X_centeredT @ X_centered)/(n-1).
cross=np.transpose(X_centered) @ X_centered
C=cross/(len(X_centered)-1)
print("The coveriance matrix: \n",C)
```

```python
# Get eigenvalues and eigen vectors using np.linalg.eigh()(symmetric matrix)
eigenvalues,eigenvectors=np.linalg.eigh(C)
print("Eigen Values: ",eigenvalues)
print("Eigen Vectors: \n",eigenvectors)
```

```python
# Sort eigen values in descending order and reorder eigenvectors accordingly.

# Sorted index
idx=np.argsort(eigenvalues)[::-1]
# Reorder both eigenvalues and eigenvectors
eigvals_sorted = eigenvalues[idx]
eigvecs_sorted = eigenvectors[:, idx]
print("Sorted Eigen Values: \n",eigvals_sorted)
print("Sorted Eigen Vectors:\n",eigvecs_sorted)
```

```python
# Compute the explained variance ratio for each componenet: 
eigval_sum=eigvals_sorted.sum()
variance_ratio=eigvals_sorted/eigval_sum
print("Explained variance ration for each component: \n",variance_ratio)
```

```python
# Project the data onto the top 2 components - report the shape of the result.
projection=X_centered @ eigenvectors[:,:2]
print("The top 2 components: \n",projection)
print("Shape: ",projection.shape)
```

# Monthly returns for 4 assets over 24 months.

```python
np.random.seed(42)
returns = np.random.randn(24, 4) * 0.05 + 0.01
# shape (24 months, 4 assets)
print("Raw\n",returns)
```

```python
# Compute the covariance matrix of asset returns(use np.cov(returns.T))
returnsT=np.transpose(returns)
cov_returns=np.cov(returnsT)
print("Covariance matrix of asset returns: \n",cov_returns)
```

```python
# Compute eigenvalues of the covariance matrix - these represent reisk componenets.
eig_vals=np.linalg.eigvalsh(cov_returns)
print("Eigen values of the covariance matrix are: \n",eig_vals)
```

```python
# Find what percentage of total risk(sum of eigenvalues) is explained by the largest eigenvalue. 
percentage_risk=(eig_vals.max()/eig_vals.sum())*100
print(f"The percentage of total risk explained by the largest eigenvalue({np.round(eig_vals.max(),5)}) is {np.round(percentage_risk,5)}%")
```

Create an equal-weight portfolio vector w=[0.25,0.25,0.25,0.25]
Compute portfolio variance and report portfolio std

```python
w=np.array([0.25,0.25,0.25,0.25])
wT=np.transpose(w)

portfolio_var=wT @ cov_returns @ w
print("Portfolio variance: ",np.round(portfolio_var,6))
print("Portfolio standard deviation: ",np.round(np.sqrt(portfolio_var),6))

```

# A simple Linear Regression solver from scratch using the Normal Equation. Dataset: house sizes and prices.

```python
# House sizes (sq ft) and prices ($1000s)
sizes  = np.array([1400, 1600, 1700, 1875, 1100,
                   1550, 2350, 2450, 1425, 1700], dtype=float)
prices = np.array([245, 312, 279, 308, 199,
                   219, 405, 324, 319, 255], dtype=float)
# np.set_printoptions(suppress=True)   # suppress scientific notation
# np.set_printoptions(precision=6)
```

```python
# Build the design matrix 'X' by stacking a column of ones(bias term) with the sizes column.
ones=np.ones(10)
X = np.column_stack([ones, sizes])
print("Matrix X:\n", X)
```

```python
# Compute weights using the Normal Equaltion: w = inv(XT X) @ XT y. where y is prices
XT=np.transpose(X)
dot_xtx=np.dot(XT,X)
dot_xty=np.dot(XT,prices)

w=np.linalg.inv(dot_xtx) @ dot_xty
print("Weights: \n",w)
```

* Intercept (bias term) = 98.25  
→ This is the baseline price (in $1000s) when the house size is 0. Of course, a house can’t be 0 sq ft, but mathematically it’s the starting point of the line.

* Slope (coefficient for size) = 0.1098  
→ For each additional square foot, the price increases by about 0.11 thousand dollars, i.e. $110 per sq ft.

```python
# Verify det(XT X)!=0 before inverting.
det_xtx=np.linalg.det(dot_xtx)
print("Determinant of dot product of XT & X: ",det_xtx)
```

Since the determinant doesn't equals to zero. So the system has unique solution.

```python
# Use the weights to predict prices for the training data: y_pred=X @ w.
y_pred=(X @ w)
y_pred_view=y_pred.reshape(-1,1)*1000
print("Predicted prices for the training data(in $): \n",y_pred_view)
```

```python
# Compute Mean Squared Error: MSE= mean((y-y_pred)^2)
MSE=np.mean(np.square(prices-y_pred))
print("Mean Squared Error: ",MSE)

RMSE=np.sqrt(MSE)
print("Root Mean Square Error: ",RMSE)
```

* Mean Squared Error (MSE) = 1366.56  
→ This is the average of the squared prediction errors. It’s in units of ($1000)^2, so less intuitive on its own.
* Root Mean Squared Error (RMSE) ≈ 37  
→ On average, the model’s predictions are off by about $37,000 compared to actual house prices.

```python
# Find which house has the largest prediction error using argmax() on the absolute error.
abs_error=prices-y_pred;
max_error_houseIdx=np.argmax(abs_error)
print(f"House no. {max_error_houseIdx+1} has the largest prediction error of roughy ${np.round(np.max(abs_error)*1000,0)}\n where actual price was ${prices[max_error_houseIdx]*1000} but system predicted ${np.round(y_pred[max_error_houseIdx]*1000,0)}")
```

```python
# Verify the Normal Equation solution matches np.linalg.solve(XT @ X,XT @ prices).
w2=np.linalg.solve(XT @ X,XT @ prices)
print("New Normal Equaltion solution: \n",w2)
print("Do (inv(XT X) @ XT prices) and np.linalg.solve(XT @ X,XT @ prices) matche?: ",np.allclose(w,w2))
```

* Both methods give the same regression weights — intercept ≈ 98.25 and slope ≈ 0.1098 — confirming that the Normal Equation solution is consistent whether you use the matrix inverse or np.linalg.solve.
