# Reshape(.reshape())

```python
import numpy as np 
arr=np.arange(1,25)
print("Orginal Array: ",arr)
```

```python
# Reshape the array so that the last row becomes: [21 22 23 24]
new_arr=arr.reshape(6,4)
print("Reshaped Array: \n",new_arr)
```

```python
# After reshaping, extract: [7 8 9 10]
print("Extracted Elements: ",new_arr[[1,1,2,2],[2,3,0,1]])
```

```python
# Reshape the array so that element 15 is located at index 2 and column index 2.
reshaped_arr=arr.reshape(4,6)
print("Reshaped Array so that arr[2][2]=15: \n",reshaped_arr)
```

```python
array=np.arange(1,37)
```

```python
# Reshape the array into a matrix and find the sum of the middle row.
matrix=array.reshape(9,4);
print("Reshaped Array: \n",matrix)
mid_rowIdx=len(matrix)//2
print("Sum of middle row: ",matrix[mid_rowIdx].sum())
```

```python
# Reshape the array again into 6 x 6 matrix and extract the border elements only.
reshaped=array.reshape(6,6)
print("Reshaped Array: \n",reshaped)

top=reshaped[0,:]
bottom=reshaped[-1,:]
left=reshaped[1:-1,0]
right=reshaped[1:-1,-1]

print("Top border element: ",top)
print("Botton border elements: ",bottom)
print("Left border element: ",left)
print("Right border elements: ",right)
```

```python
# Calculate the sum of both diagonals.
print("Diagonal element sum : ",reshaped.trace())
print("Anti-Diagonal element sum: ",np.fliplr(reshaped).trace())
```

# The data represents sales of 5 stores over 6 days.

```python
sales=np.arange(1,31)
data=sales.reshape(6,5)
print("Sales Data: \n",data)
```

```python
# Find the store with the highest total sales.
total_sales=data.sum(axis=0)
print("Total sales per store: \n",total_sales)
print(f"Store with highest sales is store with {total_sales.max()} sales")
```

```python
# Find the day with the highest overall sales.
sales_per_day=data.sum(axis=1)
print("Sales per day: ",sales_per_day)
max_day_sale=np.argmax(sales_per_day)
print(f"The day with max sales is at {max_day_sale} with sales of {sales_per_day[max_day_sale]}$")
```

```python
# Extract sales of Store 3 during the last 3 days.
store3_sales=data[-3:,2].sum()
print("Result: ",store3_sales)
```

```python
# Find the difference between the best and worst performing store.
best=total_sales.max()
worst=total_sales.min()
print("Difference between best and worst performing stores is : ",best-worst)
```
