# Mini Real-World Challenge

```python
import numpy as np 
revenue=np.arange(1000,1096)
```

A company has:
   * 8 branches
   * 12 months of revenue data

```python
# Reshape the data appropriately.
branchXmonths=revenue.reshape(8,12)
print("Revenue Table(Branch X Months):\n",branchXmonths)
```

```python
# Find the best-performing branch.
#  The best-performing branch will be the branch with the highest revenue.
gross_revenue_per_branch=branchXmonths.sum(axis=1)
print("The total revenue per branch: ",gross_revenue_per_branch)
print(f"The best-performing branch is Branch {np.argmax(gross_revenue_per_branch)+1} with total revenue of {gross_revenue_per_branch.max()}$")
```

```python
# Find the best-performing month.
gross_revenue_per_month=branchXmonths.sum(axis=0)
print("The total revenue per branch: ",gross_revenue_per_month)
print(f"The best-performing is {np.argmax(gross_revenue_per_month)+1}th Month with total revenue of {gross_revenue_per_month.max()}$")
```

```python
# Find the month with the lowest average revenue.
avg_revenue_month=branchXmonths.mean(axis=0)
print(f"The month with lowest average revenue is {np.argmin(avg_revenue_month)+1}th month with the average revenue of {avg_revenue_month.min()}")
```

```python
# Transpose the dataset and identify the month-to-month revenue trend.
monthXbranch=np.transpose(branchXmonths)

# Suffle to imitate real-world data
flat=monthXbranch.ravel()
np.random.shuffle(flat)
monthXbranch=flat.ravel().reshape(monthXbranch.shape)

total_revenue_per_month=monthXbranch.sum(axis=1)
revenue_diff_per_month=np.diff(total_revenue_per_month)
percentage_change=(revenue_diff_per_month/total_revenue_per_month[:-1])*100

print(f"Revenue dataset: \n{monthXbranch}")
print(f"Total revenue for each month across all branches: ",total_revenue_per_month)
print(f"Revenue difference across all branches within adjecent month: ",revenue_diff_per_month)
print(f"Percentage change in revenue across all branches within adjecent month: ",np.round(percentage_change,2))
```

```python
# Flatten the dataset and find the top 10 revenue values.
flat=np.sort(branchXmonths.flatten())
print("Revenues: \n",flat)

print("Top 10 revenue values: ",flat[-10:])
```

```python
# Calculate what percentage of revenues are above the overall average.
overall_avg=flat.mean()
revenues_above_overall_avg=flat[flat>overall_avg]

percentage_renvenue_above_avg=((revenues_above_overall_avg.size)/flat.size)*100

print(f"Revenues above overall average({overall_avg}): \n",revenues_above_overall_avg)
print(f"Percentage of revenues above the overall average is {np.round(percentage_renvenue_above_avg,2)}%")
```

```python
# Find the branch with the highest average revenue.
avg_revenue_branch=branchXmonths.mean(axis=1)
print(f"The branch with the highest average revenue is Branch {np.argmax(avg_revenue_branch)+1} with average revenue of {avg_revenue_branch.max()}$")
```
