#flatten()

```python
import numpy as np 
arr=np.arange(1,17).reshape(4,4)
print("Array is: \n",arr)
```

```python
# Flatten the array and find the sum of the last 5 elements.
flatten_arr=arr.flatten()
print("The sum of the last 5 elements after flatten: ",flatten_arr[-5:].sum())
```

```python
# Flatten the array and extract elements from position 5 to 10.
print("Elements from position 5 to 10 in flatten array are: ",flatten_arr[4:10])
```

```python
# Flatten the array and find the difference between the largest and smallest element.
print("Difference between largest and smallest element is ",flatten_arr.max()-flatten_arr.min())
```

# Next Module

```python
arr=np.random.randint(10,90,25).reshape(5,5)
print("The Array is : \n",arr)
```

```python
# Flatten the array and calculate the sum of elements at even indices.
flatten_arr=arr.flatten()
evenIdxEle=flatten_arr[::2]
print("Sum of elements at even index: ",evenIdxEle.sum())
```

```python
# find the product of the first 10 elements.
first_ten=flatten_arr[:10]
print("First ten elements: ",first_ten)
print("Product of first then elements: ",np.prod(first_ten))
```

```python
# Determine how many elements are divisible by both 2 and 3.
mask=(flatten_arr%2==0)&(flatten_arr%3==0)
divisible_elements=flatten_arr[mask]
print("Divisible elements: ",divisible_elements)
print("Count: ",divisible_elements.size)
```

# Next Module

```python
sales=np.random.randint(100,500,9).reshape(3,3)
print("Sales Record: \n",sales)
```

```python
flatten_arr=sales.flatten()
# Find the top 3 sales values.
top_values=3
sorted_arr=np.sort(flatten_arr)
print(f"Top {top_values} sales values are :",sorted_arr[-top_values:])
```

```python
# Find the average of all sales.
print("Average: ",np.round(flatten_arr.mean(),2))
```

```python
# Count sales greater then 140.
great_sales=flatten_arr[flatten_arr>140]
print("Sales greater than 140: ",great_sales)
print("Sales count greater than 140: ",great_sales.size)
```
