# Ravel(modifies original array)

```python
import numpy as np 
arr=np.arange(1,13).reshape(3,4)
print("Array: \n",arr)
```

```python
# Ravel the array and find the median position element.
raveled=arr.ravel()
print("Raveled Array: ",raveled) # already sorted
n=len(raveled)
if n%2==0:
    median_position=[n//2,n//2+1]
else: 
    median_position=[n//2]
print("Median Position is :",median_position)
```

```python
# Extract every third element
ele=raveled[2::3]
print("Every third elements in the raveled array : ",ele)
```

```python
# Find the sum of elements at odd positions.
print("Sum of elements at odd position: ",raveled[1::2].sum())
```

```python
arr=np.arange(1,26).reshape(5,5)
raveled_arr=arr.ravel()
print("Original array: \n",arr)
print("Raveled Array: \n",raveled_arr)
```

```python
# Calculate the sum of prime numbers.
def is_prime(n):
    if n<2:
        return False
    elif n==2:
        return True
    elif n%2==0:
        return False
    else:
        for i in range(3,int(n**0.5)+1,2):
            if n%i==0:
                return False
            return True
# Vectorization wraps a function so that it can operate elementwise operation with input of array
vec_is_prime=np.vectorize(is_prime)

prime_mask=vec_is_prime(raveled_arr)
prime_sum=raveled_arr[prime_mask].sum()
print("Prime elements are: ",raveled_arr[prime_mask])
print("Prime elemetns sum: ",prime_sum)
```

```python
# Find the average of elements greater than 10.
masked_elements=raveled_arr[raveled_arr>10]
average=masked_elements.mean()
print("Elements greater than 10: ",masked_elements)
print("Average: ",average)
```

```python
# Extract all multiples of 4 and find their sum.
masked_elements=raveled_arr[raveled_arr%4==0]
result=masked_elements.sum()
print(("Masked Elements: ",masked_elements))
print("Sum of elements that are multiples of 4: ",result)
```

```python
scores=np.arange(50,86).reshape(6,6)
print("Scores: \n",scores)
```

```python
flat=scores.ravel()
print("Raveled Array: \n",flat)
```

```python
# Find the highes 5 scores.
print("Highest 5 scores: ",flat[-5:])
```

```python
# Find the percentage of scores above 70.
n=70
ele_above_n=flat[flat>n]
count_above_n=ele_above_n.size
percentage=(count_above_n/len(flat))*100
print(f"Percentage of scores above {n} is {np.round(percentage,2)}%")
```
