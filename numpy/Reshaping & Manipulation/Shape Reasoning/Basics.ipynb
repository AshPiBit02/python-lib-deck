# Reshape, Transpose, Flatten and Ravel

Shape

```python
import numpy as np 
arr=np.arange(24)
print(arr)
```

```python
# Reshape into 2D
print("Reshaped Arrar: \n",arr.reshape(6,4))
```

```python
# Reshape into 3D array
print("Reshaped into 3D: \n",arr.reshape(4,3,2))
```

```python
# Reshape into another 2D shape
print("Reshaped in 2D again: \n",arr.reshape(2,12))
```

```python
# Flatten a 2D array
arr=np.arange(20).reshape(5,4)
print("Original Array: \n",arr)
print("Flatten Array: \n",arr.flatten())
print("Array: \n",arr) # Flatten just makes copy of array and return(original unaffected)
```

```python
# Ravel the array
print("Array Ravel: \n",arr.ravel())
```
