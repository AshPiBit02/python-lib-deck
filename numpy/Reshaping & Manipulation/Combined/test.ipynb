```python
import numpy as np 
sales=np.arange(101,137)
print("Sales data: \n",sales)
```

```python
# 1. Reshape the array into 6x6 matrix and find the sum of the third row.
sales6x6=sales.reshape(6,6)
sum_of_3R=sales6x6[2,:]
print("Reshaped Sales Array: \n",sales6x6)
print("Element of third row: \n",sum_of_3R)
print("Sum of 3rd row elements: ",sum_of_3R.sum())
```

```python
# Reshape the array into a 4x9 matrix and find the largest value in the second column.
sales4x9=sales.reshape(4,9)
second_column_ele=sales4x9[:,1]
print("Reshaped Sales Array: \n",sales4x9)
print("Elements of Second Column are : ",second_column_ele)
print("Largest values in Second column: ",second_column_ele.max())
```

```python
# Transpose the 6x6 matrix and find the sum of its last row.
transpose6x6=np.transpose(sales6x6)
tranpose_lrow=transpose6x6[-1,:]
print("Transpose Array: \n",transpose6x6)
print("Last Row of transposed Array: ",tranpose_lrow)
print("Sum of Last row of T Array: ",tranpose_lrow.sum())
```

```python
# Flatten the 6x6 matrix and extract the last 8 elements.
flat=sales6x6.flatten()
print("Last 8 element after flatten: ",flat[-8:])
```

```python
# Ravel the 6x6 matrix and find the average of elements from index 10 to 20.
ravel=sales6x6.ravel()
elements=ravel[10:21]
print("Elements from index 10 to 20 are: ",elements)
print("Average of these elements: ",elements.mean())
```

```python
sales=np.arange(101,137) # The data represents sales of 6 stores(cols) over 6 days(row)
```

```python
# Reshape appropriately and find the store with the highest total sales.
daysXstores=sales.reshape(6,6)
sales_per_store=daysXstores.sum(axis=0)
print("Sales(DaysXStores)\n:",daysXstores)
print("Sales per Store: ",sales_per_store)
print("Store with highest total sales is at index ",np.argmax(sales_per_store)," with highest sales of ",sales_per_store.max())
```

```python
# Find the day with the highest overall sales.
sales_per_day=daysXstores.sum(axis=1)
print("Sales per Day: ",sales_per_day)
print("Store with highest total sales is at index ",np.argmax(sales_per_day),"with highest sales of ",sales_per_day.max())
```

```python
# Extract the sales of Store 4 during the last three days.
store4SalesL3D=daysXstores[-3:,3]
print("Sales of store 4 during the last three days: ",store4SalesL3D)
```

```python
# Find the difference between the best and worst performing store.
best=sales_per_store.max()
worst=sales_per_store.min()
print("The difference between the best and worst performed stores: ",best-worst)
```

```python
# Transpose the data and determine which day had the highest average sales.
TdaysXsales=np.transpose(daysXstores)
highest_avg_sales_perday=TdaysXsales.sum(axis=1)
print("Transpose Sales(Day X Sales):\n",TdaysXsales)
print("Highest average sales day at index ",np.argmax(highest_avg_sales_perday)," with highest sales of ",highest_avg_sales_perday.mean())
```
