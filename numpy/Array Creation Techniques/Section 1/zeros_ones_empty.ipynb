# zeros(), ones(), & empty()

```python
# Create a 1D arry of 10 zeros.
import numpy as np 
zeros_array=np.zeros(10)
print(zeros_array)
```

```python
# Create a 3 x 4 array of ones.
ones_array=np.ones((3,4))
print(ones_array)
```

```python
# Create a 2 x 3 x 4 array filled with zeros.
ar=np.zeros((2,3,4))
print(ar)
```

```python
# Create a 5 x 5 array of ones with data type float32.
arr=np.ones((5,5),dtype=np.float32)
print(arr)
print(arr.dtype)
```

```python
# Create a 4 x4 array using emtpy().(Stores garbage values need to override)
arr=np.empty((4,4))
print(arr)
```

```python
# Create a 6 x 3 array of ones and caculate its sum.
arr=np.ones((6,3))
print(arr.sum())
```
