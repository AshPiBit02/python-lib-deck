```python
# Create a 5 x 5 identity matix and add it to a 5 x 5 matix of ones.
import numpy as np 
identitymat=np.identity(5)
onesmat=np.ones((5,5))
print("Identity Matrix: \n",identitymat)
print("Ones matrix: \n",onesmat)
print("Sum: \n",identitymat+onesmat)
```

```python
# Create two random 3 x 3 matrices and add them.
randmat1=np.random.randint(0,10,size=(3,3))
randmat2=np.random.randint(0,10,size=(3,3))
print("Matrix 1 : \n",randmat1)
print("Matrix 2 : \n",randmat2)
print("Sum : \n",randmat1+randmat2)
```

```python
# Generate a linspace() array from 0 to 100 and reshape it into a 5 x 2 matrix.
arr=np.linspace(1,100,10)
matrix=arr.reshape(5,2)
print("linspace array: \n",arr)
print("Reshped matrix : \n",matrix)
```

```python
# Create a 4 x 4 zero matrix and replace its dialognal element with 1.
zero=np.zeros((4,4),dtype=int)
np.fill_diagonal(zero,1)
print(zero)
```

```python
# Generate a random 5 x 5 matrix and calculate the sum of each row.
randmat=np.random.randint(0,10,size=(5,5))
row_sum=np.sum(randmat,axis=1)
print(randmat)
print(row_sum)
```
