# Random Arrays(np.random())

```python
# Generate 10 random values between 0 and 1.
import numpy as np 
rnd=np.random.rand(10)  # random.rand(n) function generates n floats in the range[0,1]
print(rnd)
```

```python
# Generate a 3 x 3 matrix of random values.
rnd=np.random.rand(3,3)
print(rnd)
```

```python
# Generate 20 random integers between 1 and 100.
rndint=np.random.randint(1,10,size=20)
print(rndint)
```

```python
# Generate 4 x 5 matrix of random integers between 10 and 50.
rndmat=np.random.randint(10,50,size=(4,5))
print(rndmat)
```

```python
# Generate a random array using a fixed seed and verfiy that the same output is produced repeatedly.
np.random.seed(43)
randonArr=np.random.rand(10)
print(randonArr)

np.random.seed(43)
randomArr=np.random.rand(10)
print(randomArr)

print(np.array_equal(randonArr,randomArr))
```

```python
# Generate 100 random number and find their mean.
nums=np.random.randint(1,100,size=100)
print(nums.sum())
```

```python
# Generate a 5 x 5 matrix of random integers and find the maximum and minimum values.
mat=np.random.randint(1,9,size=(5,5))
matMax=mat.max()
matMin=mat.min()
print(mat)
print("Maximum: ",matMax)
print("Minimun: ",matMin)
```

```python
# Generate 50 random integers and count how many are greater than 75.
randint=np.random.randint(0,101,50)
print("Random Integers: ",randint)

count=np.sum(randint>75)
print("Count greater than 75: ",count)
```
