# arange() & linspace()

```python
# Create an array containing number from 0 to 19 using arange().
import numpy as np 
nums=np.arange(1,20,1)
print(nums)
```

```python
# Create an array containing even numbers from 2 to 20.
even_nums=np.arange(2,22,2)
print(even_nums)
```

```python
# Create an array from 5 to 50 with a step size of 5.
nums=np.arange(5,55,5)
print(nums)
```

```python
# Create an array of decimal values from 0 to 1 with a tep size of 0.2
nums=np.arange(0,1,0.2)
print(nums)
```

```python
# Generate 50 equal spaced values between -1 to 1.
values=np.linspace(-1,1,50)
print(values)
```

```python
# Create an array fo 100 equally spaced values between 0 to 2pi.
import math
arr=np.linspace(0,2*np.pi,100)
```
