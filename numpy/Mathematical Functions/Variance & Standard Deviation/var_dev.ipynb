# Varience & Standard Deviation

```python
import numpy as np 
class1_marks=np.array([60,62,59,61,58,60])
class2_marks=np.array([10,90,15,85,20,80])
```

```python
# Which class has more consistent marks?
print("Varience of Class1 marks: ",np.var(class1_marks))
```

```python
print("Varience of Class2 marks: ",np.var(class2_marks))
```

```python
result = "Class 1 " if np.var(class1_marks)<np.var(class2_marks) else "Class 2"
print("The class with more consistent marks is ",result)
```
