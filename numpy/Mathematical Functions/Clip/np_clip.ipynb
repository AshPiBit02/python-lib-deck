# Generate 10 random floats between -5 and 5. Clip to [-1,1]. Plot the before/after using print to compare. 

```python
import numpy as np 
arr=np.random.randint(-5,5,10)
clipped_arr=np.clip(arr,-1,1)
print("Before Clipping: ",arr)
print("After Clipping: ",clipped_arr)
```

# You have a 2D image pixel array(values 0-255 but some went out of range due to brigntness adjustment). Clip it back to valid range and confirm with min() and max().

```python
image=np.random.randint(0,300,49).reshape(7,7)
print("Unclipped Image:\n",image)
clipped_image=np.clip(image,0,255)
print("Clipped Image:\n",clipped_image)
print("Max : ",clipped_image.max())
print("Min : ",clipped_image.min())
```

# Clip an array using only a lower bound(no upper limit). How do you pass the upper argument?

-> Pass arr.max() as upper argument

```python
arr=np.random.randint(-50,100,15)
print("Before Clipping: ",arr)
lwr_clipped_arr=np.clip(arr,0,arr.max())
print("After Clipping lower bound: ",lwr_clipped_arr)
```

# Compare np.clip() vs manual np.where() clipping - implement the same clipping using both and verify the outputs match.

```python
demo=np.random.randint(-20,50,25)
where_clipping=np.where(demo<0,0,np.where(demo>10,10,demo))
clip_clipping=np.clip(demo,0,10)
print("Orginal data: \n",demo)
print("Clipped using where: \n",where_clipping)
print("Clipped using clip:\n",clip_clipping)
print("Are Same?: ",np.array_equal(where_clipping,clip_clipping))
```
