# Aggregations

```python
import numpy as np 
sales=np.array([120,150,100,180,250,200])
```

```python
# What is the total revenue?
print("Total Revenue: ",sales.sum())
```

```python
# What is the average daily sale?
print("Average Daily Sales: ",sales.mean())
```

```python
# How much larger is the highest sale than the lowest sale?
print("Sales Gap Between Highest and Lowest Sale: ",sales.max()-sales.min())
```

```python
# What percentage of total revenue came from the highest-selling day?
print(f"Percentage of total revenue came for the Highest-selling day: {(sales.max()/sales.sum())*100}%")
```
