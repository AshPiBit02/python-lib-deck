# Generate 50 random salaries between 20 to 200k. Find P25, P50, P75. What percentage of values fall between P25 and P75? Verify by counting.

```python
import numpy as np 
salaries=np.random.randint(20,200,50)
P1=np.percentile(salaries,25)
P2=np.percentile(salaries,50)
P3=np.percentile(salaries,75)
print("25th Percentile(P1): ",P1)
print("50th Percentile(P2): ",P2)
print("75th Percentile(P3): ",P3)
```

```python
print("Salaries: ",salaries)
# Count values between P25 and P75
mask=(salaries>=P1)&(salaries<=P3)
count=np.sum(mask)
print("Count: ",count)

percentage=(count/50)*100
print(f"Percentage of values between P25 & P75: {percentage}%")
```

# Given a 2D array, find the 75th percentile of each column using the axis argument.

```python
import numpy as np 
arr=np.random.randint(10,100,25).reshape(5,5)
print("The 2D Array: \n",arr)
```

```python
P75=np.percentile(arr,75,axis=1)
print("The 75th percentile per column: ",P75)
```

Meaning: 80%, 88%, 65%, 94% & 51% student of column1, column2, column3, column4 and column5 have scored less than 75% percent of marks.

# What's the difference between np.percentile(arr,50) and np.median(arr)? Verify they give the same result, then check if there any edge case where they might differ?

```python
arr=np.random.randint(10,150,25)
print("Array: ",arr)
```

```python
print("Result from Percentile(P50): ",np.percentile(arr,50))
print("Result from median: ",np.median(arr))
```

Edge Cases:
 i.   Even-sized dataset with different interpolation methods
 ii.  Nan values.
 iii. Axis handling (on older version)

```python
# i
import numpy as np 
data=[1,2,3,4]
print("np.median(): ",np.median(data)) # Always average of middle two

print("np.percentile (linear): ",np.percentile(data,50,method='linear')) # default
print("np.percentile (nearest): ",np.percentile(data,50,method='nearest'))
print("np.percentile (lower): ",np.percentile(data,50,method='lower'))
print("np.percentile (higher): ",np.percentile(data,50,method='higher'))
print("np.percentile (midpoint): ",np.percentile(data,50,method='midpoint'))  # average
```

```python
# ii
data=[1,2,np.nan,4]
print("np.median: ",np.median(data)) # returns nan
print("np.nanmedian: ",np.nanmedian(data)) # Ignores NaN 
print("np.percentile: ",np.percentile(data,50)) # returns nan
```

```python
# iii. (Identical in numpy version >=1.22)
arr=np.array([[1,2],[3,4]])
print("np.median axis=0: ",np.median(arr,axis=0))
print("np.percentile axis=0: ",np.percentile(arr,50,axis=0))
```

# Using IQR, write a function remove_outliers(arr) that returns the array with outliers replaced by np.nan. 

```python
def remove_outliers(arr):
    q1=np.percentile(arr,25)
    q3=np.percentile(arr,75)
    iqr=q3-q1

    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr

    arr_out=np.where((arr<lower_bound) | (arr>upper_bound),np.nan,arr)
    return arr_out

data=np.array([10,5,16,32,90,-100,300,40])
print("Orginal data: ",data)
print("Masked data: ",remove_outliers(data))
```
