# Axis Operations

```python
import numpy as np 
marks=np.array([[80,70,90],[60,85,75],[95,88,92]]) # (student,subjects)
print(marks)
```

```python
# Find each student's total marks.
print("Total marks of each student:",marks.sum(axis=1))
```

```python
# Find each subject's total marks.
print("Total marks per subject: ",marks.sum(axis=0))
```

```python
# Which student performed best overall?
total_marks=marks.sum(axis=1)
print("Student who performed best overall is student with total marks: ",total_marks.max())
```
