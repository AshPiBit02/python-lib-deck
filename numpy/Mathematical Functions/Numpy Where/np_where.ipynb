# Given an array of temperatures. Without using any loop or if, replace: 
  * Values above 35 -> "Hot"
  * Values equal and below -> "Normal"

```python
import numpy as np 
arr=np.array([45,34,24,54,23,19,43,31,37,26])
print("Result: ",np.where(arr>35,"Hot","Normal"))
```

# Given two arrays a and b of same shape, build a result array that picks from a where a > b, else picks from b.

```python
a=np.random.randint(1,10,9).reshape(3,3)
print(a)
```

```python
b=np.random.randint(1,10,9).reshape(3,3)
print(b)
```

```python
result=np.where(a>b,a,b)
print(result)
```

# You have an array of integers. Replace all even numbers with 0 and all odd numbers with 1 using np.where().

```python
arr=np.random.randint(-10,10,16).reshape(4,4)
print(arr)
```

```python
result=np.where(arr%2==0,0,1)
print(result)
```

# Simulate a grading system: marks array -> return "Pass" if >= 40, else "Fail". Then count how many passed without using a loop.

```python
marks=np.random.randint(25,100,20)
print(marks)
```

```python
result=np.where(marks>=40,"Pass","Fail")
print(result)
```

```python
print("Number of passed students: ",np.sum(result=="Pass"))
```

# Use nested np.where() to assign greader: A(>=80), B(>=60), C(>=40),F(<40)

```python
marks=np.random.randint(15,100,30)
print(marks)
```

```python
grades=np.where(marks>=80,'A',np.where(marks>=60,'B',np.where(marks>=40,'C','F')))
print(grades)
```
