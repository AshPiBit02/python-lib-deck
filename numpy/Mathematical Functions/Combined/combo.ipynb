# Load or create a 2D dataset of 10 students x 5 test scores. Then in sequence:

```python
import numpy as np 
scores=np.random.randint(15,100,50).reshape(10,5) # row represents student and col represents test scores
print("Student Scores(Student x Scores):\n",scores)
```

1. Find the total scores according to test subjects.

```python
print("Total Scores Test: ",scores.sum(axis=1))
```

2. Clip all scores to range[30,100]

```python
print("Clipped Scores: \n",np.clip(scores,30,100))
```

3. Flag all scores below the 25th percentile using np.where().

```python
q1=np.percentile(scores,25)
print("Result: \n",np.where(scores<q1,"FLAGGED",scores))
```

4. Find the IQR accross the full dataset.

```python
q2=np.percentile(scores,75)
IQR=q2-q1
print("Interquartile Range: ",IQR)
```
