Simulate a Monte Carlo estimation of pie random points:

```python
# Generate 1,000,000 random (x,y) points where both x and y are in [-1,1]
import numpy as np 
np.random.seed(0)
points=np.random.uniform(-1,1,size=(1000000,2))
print("Points array shape: ",points.shape)
print("First five points(x,y): \n",points[:5])
```
