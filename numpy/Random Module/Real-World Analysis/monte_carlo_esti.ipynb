Simulate a Monte Carlo estimation of pie random points:

```python
# Generate 1,000,000 random (x,y) points where both x and y are in [-1,1]
import numpy as np 
np.random.seed(0)
points=np.random.uniform(-1,1,size=(1000000,2))
print("Points array shape: ",points.shape)
print("First five points(x,y): \n",points[:5])
```

```python
# A point is inside the unit circle if X^2 + Y^2 <=1.
inside_unitCirc=points[(np.square(points[:,0]))+(np.square(points[:,1]))<=1]
print("Points shape inside unit circle: ",inside_unitCirc.shape)
```

```python
inside_unitCirc=points[(np.square(points[:,0]))+(np.square(points[:,1]))<=1]
print("Points shape inside unit circle: ",inside_unitCirc.shape)
```

```python
# pi ≈ 4(points inside cirle/total points)
pi=4*(inside_unitCirc.shape[0]/points.shape[0])
print("Calulated value of pi: ",pi)
print("Numpy pi value: ",np.pi)
percentage_chng=abs((np.pi-pi)/np.pi)*100
print(f"Percentage difference between actual and calculated value: {percentage_chng:.2f}%")
```
