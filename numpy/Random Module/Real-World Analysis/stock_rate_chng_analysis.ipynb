Simulating a stock price walk for 5 assets over 252 trading days(1 year):

```python
import numpy as np 
np.random.seed(3)
start_prices = np.array([150., 280., 95., 420., 60.])
```

```python
# Generate daily returns as normal(loc=0.0005,scale=0.02,size=(252,5))
daily_returns=np.random.normal(loc=0.0005,scale=0.02,size=(252,5))
print("Daily returns shape: ",daily_returns.shape)
```

```python
# Compute comulative price series.
prices=start_prices*np.cumprod(1+daily_returns,axis=0)
print(prices)
```

```python
# Find the final price of each asset after 252 days.
final_price=prices[-1]
print("Final price of each asset after 252 days:\n",final_price)
```

```python
# Find which asset had the highest single-day percentage gain across the entire simulation on the flattend returns.
max_gain_index = np.argmax(daily_returns)
day,asset=divmod(max_gain_index,daily_returns.shape[1])
max_gain_value=daily_returns[day,asset]
print(f"The highest single-day percentage gain asset was Asset{asset+1} with percentage gain of {max_gain_value*100:.2f}%")
```

```python
# Find the worst day overall(largest single_day drop).
min_drop_index = np.argmin(daily_returns)          # flat index of worst return
day, asset = divmod(min_drop_index, daily_returns.shape[1])
min_drop_value = daily_returns[day, asset]
print(f"The worst day overall is Day{day+1} with largest single-day drop of {min_drop_value*100:.2f}%")
```
