# Building a bootstrapped confidence interval for the mean of a dataset.
(This is a core statistical technique used when you con't assume a distribution.)

```python
import  numpy as np 
np.random.seed(0)
# simulate exam scores for 80 students
population=np.random.normal(loc=68,scale=15,size=80).clip(0,100)
print("Polpulation shape: ",population.shape)
```

```python
# Compute the true sample mean of 'population'.
true_sample_mean=population.mean()
print(f"True Sample Mean: {true_sample_mean:.2f}")
```

Run 10,000 bootstrap iterations:

```python
# Each iteration: sample 80 values with replacement from population using choice().
# Store the mean of each bootstrap sample -reshult shape:(10000,)
# Without a python loop using a single choice() call with size(10000,80) then mean along axis=1.

bootstrap_samples=np.random.choice(population,size=(10000,80),replace=True)
bootstrap_means = bootstrap_samples.mean(axis=1)
print("Bootstrap Samples shape: ",bootstrap_samples.shape)
print("First 5 bootstrap means:", bootstrap_means[:5])
```

```python
# Compute the 95% confidence interval using np.percentile() on the bootstrap means- [2.5, 97.5]
ci_lower, ci_upper = np.percentile(bootstrap_means, [2.5, 97.5])
print("95% CI:", ci_lower, ci_upper)
```

```python
# Check if the true sample mean fall within the confidence interval.
print("Inside Confidence Interval: ",ci_lower<=true_sample_mean<=ci_upper)
```

```python
# Reshape the 10,000 boostrap means into (100,100) and compute the std of each row - report the min and max std.
reshaped_bootstrap_mean=bootstrap_means.reshape(100,100)
print("Bootstrap Mean Shape: ",reshaped_bootstrap_mean.shape)

bootstrap_means_std=np.std(reshaped_bootstrap_mean,axis=1)
print("Max Std of bootstrap mean: ",bootstrap_means.max())
print("Min Std of bootstrap mean: ",bootstrap_means.min())
```
