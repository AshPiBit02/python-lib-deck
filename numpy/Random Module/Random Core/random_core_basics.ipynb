Simulate a biased coin - heads appears 70% of the time, tails 30%.

```python
import numpy as np 
flips=np.random.choice(["H","T"],size=1000,p=[0.7,0.3])

# Count heads and tails using Boolean conditions
heads=np.sum(flips=="H")
tails=np.sum(flips=="T")

# Report results
print("Heads: ",heads)
print("Tails: ",tails)

# Check closeness to expected ratio
ratio_heads = heads / 1000
ratio_tails = tails / 1000
print("Heads ratio:", ratio_heads)
print("Tails ratio:", ratio_tails)

if abs(ratio_heads - 0.7) < 0.05 and abs(ratio_tails - 0.3) < 0.05:
    print("✅ The actual ratio is close to the expected 70/30 split.")
else:
    print("⚠️ The ratio deviates noticeably from 70/30.")
```

A dataset of 50 students scores:

```python
np.random.seed(21)
scores = np.random.normal(loc=72, scale=12, size=50).clip(0, 100).astype(int)
# About two-third random values will be close to 72(with magnitude above or below by 12)
print(scores)
```

```python
# Shuffle the scores and pick the first 10 as a "random sample".
np.random.shuffle(scores)
rand_samples=scores[:11]
print("First 10 random samples: ",rand_samples)
```

```python
# From that sample, find the min, max & mean.
print("Minimum value: ",rand_samples.min())
print("Maximum value: ",rand_samples.max())
print("Mean value: ",rand_samples.mean())
```

```python
# Repeat the process using choice() with replace=False instead of shuffle.
rand_samples=np.random.choice(scores,size=10,replace=False)
print("Random Samples: ",rand_samples)
print("Minimum value: ",rand_samples.min())
print("Maximum value: ",rand_samples.max())
print("Mean value: ",rand_samples.mean())
```

Generate a 6x6 grid of random integers between 1 and 100.

```python
data=np.random.randint(1,101,size=(6,6))
print("Generated 6x6 random integers grid: \n",data)
```

```python
# Replace all values above 80 with the value 99
data=np.where(data<20,1,data)
data=np.where(data>80,99,data)
print("Clipped data: \n",data)
```

```python
# Report how many cells are above 80 and how many are below 20.
above_80=np.sum(data>80)
below_20=np.sum(data<20)
print("Cells count above 80: ",above_80)
print("Cells count below 20: ",below_20)
```

Simulate rolling two 6-sided dice 10,000 times and build a frequency table of the sums.

```python
np.random.seed(0)
```

```python
# Generate both dice rolls in one step as a (10000,2) array.
dice_rolls=np.random.randint(1,7,size=(10000,2))
print("Results: \n",dice_rolls)
```

```python
# Sum along axis=1 to get total scores(range 2-12).
sums=dice_rolls.sum(axis=1)
print("Sums: ",sums)
```

```python
# Build freqeuncy table.
unique_sums,counts=np.unique(sums,return_counts=True)
freq_table=np.column_stack([unique_sums,counts])
print("Frequency Table: \n",freq_table)
```

```python
# Find most frequent sum.
most_frequent_idx=np.argmax(counts)
most_frequent_sum=unique_sums[most_frequent_idx]
print("The most frequent sum: ",most_frequent_sum)
```

A sales dataset for 5 products over 30 days.

```python
np.random.seed(7)
sales = np.random.randint(50, 500, size=(30, 5))
# rows = days, cols = products
```

```python
# Spilt into 3 equal block of days.
block1,block2,block3=np.vsplit(sales,3)
print("First 10 day block: \n",block1)
print("Second 10 day block: \n",block2)
print("Third 10 day block: \n",block3)
```

```python
# For each block, compute total sales per product.
Tsales_block1=np.sum(block1,axis=0)
print("Total sales per product in Block1: ",Tsales_block1)
Tsales_block2=np.sum(block2,axis=0)
print("Total sales per product in Block2: ",Tsales_block2)
Tsales_block3=np.sum(block3,axis=0)
print("Total sales per product in Block3: ",Tsales_block3)
```

```python
# Stack the 3 totals into a (3,5) summary.
summary=np.column_stack([Tsales_block1,Tsales_block2,Tsales_block3])
print("Summary: \n",summary)
```

```python
# Shuffle the rows of the summary to simulate randomized reporting order.
np.random.shuffle(summary)
print("Shuffled summary: \n",summary)
```

```python
# Identify which product had the highest total across all 30 days using argmax.
highestT_sales=np.sum(sales,axis=0)
highestS_prod_idx=np.argmax(highestT_sales)
print(f"The product that had the highest total sales across all 30 days was Product{highestS_prod_idx+1} with sales of ${highestT_sales[highestS_prod_idx]}.")
```

Generate a dataset of 200 samples from a normal distribution(mean=50,std=10) and 200 samples from a uniform distribution(range 30-70).

```python
dataset1=np.random.normal(loc=50,scale=10,size=200).astype(float)
dataset2=np.random.uniform(30,71,size=200)
print("Dataset1 shape: ",dataset1.shape)
print("Dataset2 shape: ",dataset2.shape)
```

```python
# Combine them into one array of 400 samples.
dataset3=np.hstack([dataset1,dataset2])
print("Dataset3 shape: ",dataset3.shape)
```

```python
# Compute mean,std,min,max,25th, and 75th percentile of the combined data.
mean=dataset3.mean()
std=np.std(dataset3)
min=dataset3.min()
max=dataset3.max()
Q1=np.percentile(dataset3,25)
Q3=np.percentile(dataset3,75)
print("Mean: ",mean)
print("Standara Deviation: ",std)
print("Minimum value: ",min)
print("Maximum value: ",max)
print("25th Percentile: ",Q1)
print("75th Percentile: ",Q3)
```

```python
# Use Boolean indexing to find what percentage of values fall within one std of the mean.
within_range=dataset3[(dataset3 >= mean-std) & (dataset3 <= mean+std)]
percentage=len(within_range)/len(dataset3)*100
print(F"{percentage:.2f}% of values fall within one std of the mean.")
```

```python
# Reshape the combined array into (40,10) and compute the mean of each row.
reshaped_dataset=dataset3.reshape(40,10)
print("Dataset shape: ",reshaped_dataset.shape)
mean_row=reshaped_dataset.mean(axis=1)
print("Mean of each row: \n",mean_row)
```

A train/validation/test split manually:

```python
np.random.seed(42)
data = np.random.randn(100, 5)
labels = np.random.randint(0, 3, size=100)
```

```python
# Use permutation() to get a shuffled index array.
indices=np.random.permutation(len(data))
print("Indices: ",indices)
```

```python
# Use that index array with fancy indexing to reorder both data and labels consistently.
shuffled_data=data[indices]
shuffled_labels=labels[indices]
```

```python
# Split into 70% train,15% validation, 15% test using split() at the right indices.
# Split sizes
n_train = int(0.7 * len(data))   # 70 samples
n_val   = int(0.15 * len(data))  # 15 samples
n_test  = len(data) - n_train - n_val  # 15 samples

# Train/Validation/Test splits
train_data, val_data, test_data = np.split(shuffled_data, [n_train, n_train + n_val])
train_labels, val_labels, test_labels = np.split(shuffled_labels, [n_train, n_train + n_val])

print("Train shape:", train_data.shape, train_labels.shape)
print("Validation shape:", val_data.shape, val_labels.shape)
print("Test shape:", test_data.shape, test_labels.shape)
```

```python
# Verify all three split together reconstruct the full shuffled dataset using vstack.
reconstructed_data=np.vstack([train_data,val_data,test_data])
print("Is reconstructed data and full shuffled data equivalent? :",np.allclose(reconstructed_data,shuffled_data))
```
