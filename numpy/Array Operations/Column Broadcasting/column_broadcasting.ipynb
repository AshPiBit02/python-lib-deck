# Column Broadcasting

```python
import numpy as np 
matrix=np.arange(1,10,1).reshape(3,3)
col=np.array([[10],[20],[30]])
```

```python
# Add column to matrix.
print(matrix+col)
```

```python
# Multiply matrix by col.
print(matrix * col)
```

```python
# Subtract col from matrix.
print(matrix-col)
```

```python
# Divide matrix by col.
print(matrix/col)
```

Explain why Broadcasting works here.
-> For broadcasting compatibility is required and to be compatible either both arrays must be of same dimesion or one of the dimesion is 1.

# For the compatibility of 1D array either they have the same length or one of the has length 1
