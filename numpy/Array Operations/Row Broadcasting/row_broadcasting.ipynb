# Row Broadcasting

```python
import numpy as np 
matrix=np.arange(1,10,1).reshape(3,3)
row=np.array([10,20,30])
```

```python
# Add row to matrix.
print(matrix+row)
```

```python
# Subtract row from matrix.
print(matrix-row)
```

```python
# Multiply matrix by row
print(matrix*row)
```

```python
# Divide matrix by row.
print(matrix/row)
```
