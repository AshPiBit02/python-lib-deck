# Challenge Questions

```python
import numpy as np 
```

```python
# Create two random arrays of shape(5,5) and perform all arithemtic operations.
arr1=np.random.randint(1,50,25).reshape(5,5)
print(arr1)
```

```python
arr2=np.random.randint(1,50,25).reshape(5,5)
print(arr2)
```

```python
print("Sum: \n",arr1+arr2)
print("Difference: \n",arr1-arr2)
print("Product: \n",arr1*arr2)
print("Division: \n",arr1-arr2)
print("Modulo: \n",arr1%arr2)
```

```python
print("Sum of total elements of arr1: ",arr1.sum())
print("Sum of total elements of arr2: ",arr2.sum())
```

```python
# Create a matrix and add row-wise bonuses using broadcasting.
matrix=np.arange(1,10,1).reshape(3,3)
print(matrix)
```

```python
row_wise=np.array([[10],[20],[30]])
print("Row wise addition :\n",row_wise+matrix)
```

```python
# Create a matrix and add column-wise bonuses using broadcasting.
coln_wise=np.array([10,20,30])
print("Column wise addition: \n",matrix+coln_wise)
```

```python
# Normalize an array by dividing each element by the maximum value.
print("Normalized array: \n",np.round(matrix/matrix.max(),2))
```

```python
# Compute the percentage increase between two arrays element-wise.
array1=np.arange(1,10,1).reshape(3,3)
array2=np.arange(10,19,1).reshape(3,3)
print("Array 1: \n",array1)
print("Array 2: \n",array2)
```

```python
array3=((array2-array1)/array2)*100
print("Element-wise % increase\n",array3)
```

```python
# Create a 4x4 matrix and multiply each row by a different value using broadcasting.
matrix4=np.arange(1,17,1).reshape(4,4)
print(matrix4)
```

```python
row_multiplier=np.array([[2],[4],[6],[8]])
print("Row-wise multiplication 2R1,4R2,6R3,8R4:\n",matrix4*row_multiplier)
```

```python
# Without loops, calculated the square and cube for all elements in a random array.
random_arr=np.random.randint(1,9,16).reshape(4,4)
print("Random array: \n",random_arr)
```

```python
print("Squared random array: \n",np.square(random_arr))
```

```python
print("Cube element of random array:\n",np.power(random_arr,3))
```
