# Arithmetic Operations on 2D Arrays

```python
import numpy as np 
a=np.arange(1,7,1).reshape(2,3)
b=np.arange(10,61,10).reshape(2,3)
```

```python
# Add a and b.
print(a+b)
```

```python
# Subtract a and b.
print(b-a)
```

```python
# Multiply a and b.
print(a*b)
```

```python
# Divide b by a.
print(b/a )
```

```python
# Find the square fo every element in a.
print(a**2)
```

```python
# Add 5 to every element of a.
print(a+5)
```

```python
# Multiply every element of b by 3.
print(b*3)
```

```python
# Computer (a+b)/2
print((a+b)/2)
```
