# Arithmetic Operations on 1D Arrays

```python
import numpy as np 
a=np.array([10,20,30,40,50])
b=np.array([1,2,3,4,5])
```

```python
# Add a and b.
print(a+b)
```

```python
# Subtract b from a
print(a-b)
```

```python
# Multiply a and b.
print(a*b)
```

```python
# Divide a by b.
print(a/b)
```

```python
# find a%b.
print(a%b)
```

```python
# Calculate a ** b
print(a**b)
```

```python
# Multiply every element of a by 10.
print(a*10)
```

```python
# Add 100 to every element of b.
print(b+100)
```

```python
# Divide all elements of a by 2.
print(a/2)
```

```python
# Compute (a+b)/2
print((a+b)/2)
```
