# Element-wise Operations

```python
import numpy as np 
arr=np.array([2,4,6,8,10])
```

```python
# Square every element.
print(np.square(arr))
```

```python
# Cube every element.
print(np.power(arr,3))
```

```python
# Find the reciprocal of each element.
print(np.reciprocal(arr.astype(float)))
```

```python
# Add 1 to every element
print(arr+1)
```

```python
# Multiply each element by itself.
print(arr*arr)
```

```python
# Calculate arr + arr
print(arr+arr)
```

```python
# Create a new array where each element is mulitplied by 100.
new_arr=arr*100
print(new_arr)
```
