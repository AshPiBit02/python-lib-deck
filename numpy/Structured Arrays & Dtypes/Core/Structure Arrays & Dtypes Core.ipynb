A dataset stores student information:

```python
import numpy as np 
students = np.array([
    (1, 78.5, 85),
    (2, 91.0, 88),
    (3, 67.5, 72),
    (4, 88.0, 90)
], dtype=[('id','i4'), ('marks','f4'), ('attendance','i4')])
```

```python
# Find the student ID with highest marks.
topper_id=students['id'][students['marks'].argmax()]
print("Topper ID: ",topper_id)
```

```python
# Find the average attendance.
avg_attendance=students['attendance'].mean()
print("Average Attendance: ",avg_attendance)
```

```python
# Count students with marks above 80.
student_marks_above80=np.sum(students['marks']>80)
print("Students count having marks above 80: ",student_marks_above80)
```

Field Logic + Filtering

```python
sales = np.array([
    ('Laptop', 75000, 5),
    ('Phone', 30000, 10),
    ('Tablet', 45000, 7),
    ('Monitor', 20000, 12)
], dtype=[('item','U10'), ('price','i4'), ('quantity','i4')])
```

```python
# Find total revenue for each item.
total_rev=sales['price']*sales['quantity']
print("Total Revenue:")
print("Item        Total Revenue")
for i,r in zip(sales['item'],total_rev):
    print(f"{i}      ${r}")
```

```python
# Identify the item contributing the highest revenue.
high_rev_item=sales['item'][total_rev.argmax()]
print("Item contributing the highest revenue: ",high_rev_item)
```

```python
# Count items with revenue above 300_000.
count=np.sum(total_rev>300_000)
print(count)
```
