A dataset stores student information:

```python
import numpy as np 
students = np.array([
    (1, 78.5, 85),
    (2, 91.0, 88),
    (3, 67.5, 72),
    (4, 88.0, 90)
], dtype=[('id','i4'), ('marks','f4'), ('attendance','i4')])
```

```python
# Find the student ID with highest marks.
topper_id=students['id'][students['marks'].argmax()]
print("Topper ID: ",topper_id)
```

```python
# Find the average attendance.
avg_attendance=students['attendance'].mean()
print("Average Attendance: ",avg_attendance)
```

```python
# Count students with marks above 80.
student_marks_above80=np.sum(students['marks']>80)
print("Students count having marks above 80: ",student_marks_above80)
```

Field Logic + Filtering

```python
sales = np.array([
    ('Laptop', 75000, 5),
    ('Phone', 30000, 10),
    ('Tablet', 45000, 7),
    ('Monitor', 20000, 12)
], dtype=[('item','U10'), ('price','i4'), ('quantity','i4')])
```

```python
# Find total revenue for each item.
total_rev=sales['price']*sales['quantity']
print("Total Revenue:")
print("Item        Total Revenue")
for i,r in zip(sales['item'],total_rev):
    print(f"{i}      ${r}")
```

```python
# Identify the item contributing the highest revenue.
high_rev_item=sales['item'][total_rev.argmax()]
print("Item contributing the highest revenue: ",high_rev_item)
```

```python
# Count items with revenue above 300_000.
count=np.sum(total_rev>300_000)
print(count)
```

```python
weather = np.array([
    ('Kathmandu', 25, 60),
    ('Pokhara', 28, 70),
    ('Biratnagar', 32, 55),
    ('Lalitpur', 24, 65)
], dtype=[('city','U12'), ('temp','i4'), ('humidity','i4')])
```

```python
# Find the hottest city.
hottest_city=weather['city'][weather['temp'].argmax()]
print("Hottest City: ",hottest_city)
```

```python
# Find city with lowest humidity.
lowest_humidity_city=weather['city'][weather['humidity'].argmin()]
print("Lowest Humidity city: ",lowest_humidity_city)
```

```python
# Count cities where temperature > 26 AND humidity < 65.
count=np.sum((weather['temp']>26) & (weather['humidity']<65))
print("City count: ",count)
```

```python
students = np.array([
    (1, 80, 85),
    (2, 90, 70),
    (3, 75, 95),
    (4, 85, 88),
    (5, 60, 92)
], dtype=[('id','i4'), ('math','i4'), ('science','i4')])
```

```python
# Compute total score for each student.
total_score=students['math']+students['science']
print("Total Scores: ",total_score)
```

```python
# Rank Students based on total scores.
ranks=np.argsort(total_score)[::-1]
ranked_students=students[ranks]
print("Ranked student according to total scores: ",ranked_students['id'])
```

```python
# Find student with highest improvement gap between subjects.
improvemet_gap=abs(students['math']-students['science'])
student_with_highest_imp_marks=students['id'][improvemet_gap.argmax()]
print("Student with highest imporvement gap between subjects is student with student ID:",student_with_highest_imp_marks, "with gap of",improvemet_gap.max())
```

```python
# Count students show both subjects are above average.
math_avg=students['math'].mean()
science_avg=students['science'].mean()
count=np.sum((students['math']>math_avg) & (students['science']>science_avg))
print("Student count: ",count)
```
