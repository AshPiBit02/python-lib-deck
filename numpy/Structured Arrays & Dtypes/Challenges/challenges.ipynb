Real Data Thinking

```python
import numpy as np
data = np.array([
    (1, 22, 50000, 3),
    (2, 28, 75000, 5),
    (3, 35, 60000, 2),
    (4, 40, 90000, 8),
    (5, 26, 65000, 4)
], dtype=[('id','i4'), ('age','i4'), ('salary','i4'), ('experience','i4')])
```

```python
# Compute salary per year of experience.
salary_per_experience=data['salary']*data['experience']
print("Salaray per year of experience: ")
print("ID    Net Salary")
for id,net_salary in zip(data['id'],salary_per_experience):
    print(f"{id}       {net_salary}")
```

```python
# Find the most efficient empolyee(salary per experience highest).
most_efficient_emp=data['id'][salary_per_experience.argmax()]
print("Most efficient employee: ",most_efficient_emp)
```

```python
# Find average salary of empolyees older than 30.
avg_sal_over30=data['salary'][data['age']>30].mean()
print("Average salary of employees older than 30: $",avg_sal_over30)
```

```python
# Rank employees by efficiency.
ranks=np.argsort(salary_per_experience)[::-1]
ranked_employees=data['id'][ranks]
print("Ranked employees by efficiency: ",ranked_employees)
```

```python
# Identify employees above both average salary and average experience.
avg_exp=data['experience'].mean()
avg_sal=data['salary'].mean()
filtered_empoyees=data['id'][(data['salary']>avg_sal) & (data['experience']>avg_exp)]
print("Empoyess above both average salary and average experience: ",filtered_empoyees)
```

Advanced Structured Thinking

```python
students=np.array([
    ('A',85,90,80),
    ('B',70,60,75),
    ('C',95,88,92),
    ('D',65,72,68)
],dtype=[('name','U5'),('math','i4'),('physics','i4'),('chemistry','i4')])
```

```python
# Compute total and average marks per students.
total_marks=students['math']+students['physics']+students['chemistry']
avg_marks=total_marks/3
print("Name    Total marks       Average marks")
for n,t,a in zip(students['name'],total_marks,avg_marks):
    print(f"{n}          {t}              {a:.2f}")
```

```python
# Find topper.
print(f"The topper is {students['name'][total_marks.argmax()]}")
```

```python
# Find most consistent student.
most_consistent=np.var(students['math'])
```

```python
# Identify subject-wise topper.
toppers=np.array([students['name'][students['math'].argmax()],students['name'][students['physics'].argmax()],students['name'][students['chemistry'].argmax()]])
subjects=np.array(["Math","Physics","Chemistry"])
print("Subject      Topper")
for s,t in zip(subjects,toppers):
    print(f"{s}           {t}")
```

```python
# Count students above class average in all subjects.
class_avg=np.hstack((students['math'],students['physics'],students['chemistry'])).mean()
count=np.sum((students['math']>class_avg)&(students['physics']>class_avg)&(students['chemistry']>class_avg))
print("Student count: ",count)
```

Real-World Dataset Thinking

```python
sales = np.array([
    ('StoreA', 1200, 30),
    ('StoreB', 1500, 20),
    ('StoreC', 1000, 25),
    ('StoreD', 1800, 15)
], dtype=[('store','U10'), ('revenue','i4'), ('cost','i4')])
```

```python
# Compute profit.
profit=sales['revenue']-sales['cost']
print("Store    Profit")
print('-'*20)
for s,p in zip(sales['store'],profit):
    print(f"{s}    ${p}")
```

```python
# Compute profit margins.
profit_margins=(profit/sales['revenue'])*100
print("Store     Profit       Revenue       Profit Margin")
print('-'*50)
for s,p,r,pm in zip(sales['store'],profit,sales['revenue'],profit_margins):
    print(f"{s}       {p}       {r}         {pm:.2f}%")
```

```python
# Rank stores by profitability.
ranks=np.argsort(profit)
print("Stores ranked by profitability: ",sales['store'][ranks])
```

```python
# Identify most efficient store.
most_eff_store=np.argmax(profit_margins)
print("Most efficient store is ",sales['store'][most_eff_store])
```

```python
# Find stores above average profit margin.
avg_prof_margin=profit_margins.mean()
filtered=np.where(profit_margins>avg_prof_margin)
print(f"Average Profit Margin: {avg_prof_margin:.2f}")
print("Store      Profit Margin")
print('-'*30)
for s,pm in zip(sales['store'][filtered],profit_margins[filtered]):
    print(f"{s}           {pm:.2f}%")
```
