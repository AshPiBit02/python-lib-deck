Real Data Thinking

```python
import numpy as np
data = np.array([
    (1, 22, 50000, 3),
    (2, 28, 75000, 5),
    (3, 35, 60000, 2),
    (4, 40, 90000, 8),
    (5, 26, 65000, 4)
], dtype=[('id','i4'), ('age','i4'), ('salary','i4'), ('experience','i4')])
```

```python
# Compute salary per year of experience.
salary_per_experience=data['salary']*data['experience']
print("Salaray per year of experience: ")
print("ID    Net Salary")
for id,net_salary in zip(data['id'],salary_per_experience):
    print(f"{id}       {net_salary}")
```

```python
# Find the most efficient empolyee(salary per experience highest).
most_efficient_emp=data['id'][salary_per_experience.argmax()]
print("Most efficient employee: ",most_efficient_emp)
```

```python
# Find average salary of empolyees older than 30.
avg_sal_over30=data['salary'][data['age']>30].mean()
print("Average salary of employees older than 30: $",avg_sal_over30)
```

```python
# Rank employees by efficiency.
ranks=np.argsort(salary_per_experience)[::-1]
ranked_employees=data['id'][ranks]
print("Ranked employees by efficiency: ",ranked_employees)
```

```python
# Identify employees above both average salary and average experience.
avg_exp=data['experience'].mean()
avg_sal=data['salary'].mean()
filtered_empoyees=data['id'][(data['salary']>avg_sal) & (data['experience']>avg_exp)]
print("Empoyess above both average salary and average experience: ",filtered_empoyees)
```
