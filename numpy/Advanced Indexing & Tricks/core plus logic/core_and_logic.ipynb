Advanced Argmax & Argmin

```python
import numpy as np 
sales=np.array([
    [120,140,160],
    [180,110,170],
    [130,200,150]
])
```

```python
# Identify the position of the highest sale.
max_coords=np.unravel_index(sales.argmax(),sales.shape)
print("Max sale: ",sales[max_coords])
print("Position(row,col): ",max_coords)

# Identify the position of the lowest sale.
min_coords=np.unravel_index(sales.argmin(),sales.shape)
print("Min sale: ",sales[min_coords])
print("Position: ",min_coords)

# Calulate the difference between them.
print("Difference: ",sales[max_coords]-sales[min_coords])
```

```python
prices=np.array([150,200,250,300,350,400,450])
```

```python
# Extract values located at positions: 0,2,4,6
values_at=[0,2,4,6]
selected_price=prices[values_at]
print("Required values: ",selected_price)

# Determine average selected price.
print("Average of selected price: $",selected_price.mean())

# Determine Highest selected price.
print("Highest Selected price: $",selected_price.max())

# Determine difference between highest and lowest selected price.
print("Difference between highest and lowest selected price: $",selected_price.max()-selected_price.min())
```

# Put


```python
inventory=np.arange(10,61,10)
print("Inventory: ",inventory)
```

```python
# Position 1,2,5 must be replaced by 100,200,300.
positions=[1,2,5]
values=[100,200,300]
new_inventory=inventory.copy()
np.put(new_inventory,positions,values)
print("New inventory: ",new_inventory)
```

```python
# After replcement: 
   # 1. what is the inventory total?
print("Inventory total after replacement: $",new_inventory.sum())
   # 2. Which position contains the largest values?
print("Position with largest value: ",np.argmax(new_inventory))
   # 3. What percentage increase occurred in total inventory?
print(f"Percentage increase in total inventory: {(new_inventory.sum()-inventory.sum())/inventory.sum()*100:.2f} %")
   
```

# Clip


```python
marks=np.array([25,45,65,85,105])
```

```python
# Marks below to becomes 40 & marks above 100 becomes 100.
clipped_marks=np.clip(marks,40,100)
print("Clipped marks: ",clipped_marks)

# New average
print("New Average: ",clipped_marks.mean())

# Total marks removed/added due to clipping.
print("Total marks added/removed due to clipping: ",abs(marks-clipped_marks).sum())
```

```python
data=np.arange(1,37).reshape(6,6)
print("Original data: \n",data)
```

```python
# Extract only rows having odd indices.
row_odd_idx=data[1::2,:]
print("Rows having odd indices: \n",row_odd_idx)
```

```python
# From the extracted result, keep only columns having even indices.
even_cols=row_odd_idx[:,::2]
print("Columns having even indices: \n",even_cols)
```

```python
# Find the largest value in the final matrix.
print("Largest value in the final matrix: ",even_cols.max())
```

```python
# Determine its position within the extracted matrix.
coords=np.unravel_index(even_cols.argmax(),even_cols.shape)
print(f"Position of largest value in resulting matrix: ({coords[0]},{coords[1]})")
```

```python
profits=np.array([
    420,380,510,470,
    390,600,450,530
])
```

```python
# Rank companies from highest profit to lowest.
rank_idx=np.argsort(profits)[::-1]
ranked_profits=profits[rank_idx]
print("Ranked Companies: \n",ranked_profits)
```

```python
# Determine the top three companies.
top_three_comp=rank_idx[:3]
print("Top three companies are: ",top_three_comp+1)
```

```python
# Find the difference between profits of the best and third-best company.
profit=profits[rank_idx[0]]-profits[rank_idx[2]]
print("Difference between profits of the best and third-best company: ",profit)
```

```python
matrix=np.arange(1,26).reshape(5,5)
print("Matrix: \n",matrix)
```

```python
# Calculate diagonal sum.
diagonal_sum=np.trace(matrix)
print("The diagonal sum: ",diagonal_sum)
```

```python
# Calulate upper-triangular sum.
upp_tri_sum=np.triu(matrix,k=1).sum()
print("Upper-triangular sum: ",upp_tri_sum)
```

```python
# Calculate lower-triangle sum.
lwr_tri_sum=np.tril(matrix,k=-1).sum()
print("Lower triangle sum: ",lwr_tri_sum)
```

```python
# Verify: upper+lower - diagonal= total matrix sum.
total_sum=matrix.sum()
LHS=upp_tri_sum+lwr_tri_sum+diagonal_sum
print("Is Upper+Lower-Diagonal equal to Total matrix sum?: ",np.allclose(LHS,total_sum))
```

```python
sensor=np.array([
    [12,75,130],
    [55,95,210],
    [40,180,300]
])
```

```python
# Find corrected matrix(clip min=50,max=200)
clipped_data=np.clip(sensor,50,200)
print("Corrected matrix: \n",clipped_data)
```

```python
# Find new maximum.
new_max=clipped_data.max()
print("New maximum: ",new_max)
```

```python
# Find the total reduction caused by clipping.
print("Total reduction caused by clipping: ",sensor.sum()-clipped_data.sum())
```
