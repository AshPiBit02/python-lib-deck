Advanced Slicing

```python
import numpy as np 
arr=np.arange(1,26).reshape(5,5)
print("Original matrix: \n",arr)
```

```python
# Without modifying the original array, extract every second row and every second column. 
matrix=arr[1::2,1::2]
print("Resulting matrix: \n",matrix)
```

```python
# find the sum of all corner elements of the extracted matrix.
print("Sum of corner elements of the matrix: ",matrix.sum()) #since all the elements are corner element
```

Argsort

```python
scores=np.array([78,92,65,88,95,70])
```

```python
# Determine the ranking of students from lowest score to highest.
indexes=np.argsort(scores)
print("Scores: \n",scores[indexes])
```

```python
# Which student index would appear third in the sorted order?
print(f"Student whose index would appear third in the sorted order is Student{indexes[2]+1}")
```

```python
# Which student has the second-highest scores?
print(f"The student with second-highest score is Student{indexes[-2]+1} with score of {scores[indexes[-2]]}")
```

Nonzero

```python
data=np.array([0,12,0,8,5,0,14,7])
```

```python
# Find the position containing actaul values.
print("Position containing actual values: ",np.nonzero(data))
# What is the sum of all non-zero values?
print("Sum of non-zero values: ",data[np.nonzero(data)].sum())
# What percentage of entries are non-zero?
non_zero=data[np.nonzero(data)].size
percentage=non_zero/data.size*100
print(f"Percentage of non-zero entries: {percentage:.2f} %")
```

Diag,Triu,Tril

```python
A=np.arange(1,17).reshape(4,4)
print("A: \n",A)
```

```python
# Find the sum of the main diagonal.
print("The sum of main diagonal: ",np.diag(A).sum())
# Find the sum of elemenets above the diagonal.
print("The sum of elements above the diagonal: ",np.triu(A,k=1).sum())
# Find the sum of elements below the diagonal.
print("The sum of elements below the diagonal: ",np.tril(A,k=-1).sum())
```
