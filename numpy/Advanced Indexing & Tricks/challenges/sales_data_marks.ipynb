Challenge 1

```python
import numpy as np 
sales = np.array([
    [120,140,160,180],
    [100,220,130,150],
    [190,170,210,110],
    [250,90,180,200]
])
```

```python
# Find the top five sales values.
top_indices=np.argsort(sales,axis=None)[::-1]
top_5indices=top_indices[:5]

coords=[np.unravel_index(idx,sales.shape) for idx in top_5indices]
top5_values=sales.flatten()[top_5indices]
print("Top 5 sales values: ",top5_values)

# Determine their positions.
for val,pos in zip(top5_values,coords):
    print(f"Value: {val}, Position: ({pos[0]},{pos[1]})")
```

```python
# Calculate the sum of all values above overall average.
overall_avg=sales.mean()
flat_sales=sales.flatten()
sum_above_avg=flat_sales[flat_sales>overall_avg].sum()
print("Sum of all values above overall average: ",sum_above_avg)
```

```python
# Find the branch contributing the highest revenue.
revenue_per_branch=sales.sum(axis=0) # columns as branch
print(f"Branch contributing the highest revenue is Branch{np.argmax(revenue_per_branch)+1} with revenue of ${revenue_per_branch[np.argmax(revenue_per_branch)]}.")
```

```python
# Find the revenue difference between best and worst branch.
print(f"Revenue diff between best and worst branch is of ${revenue_per_branch[np.argmax(revenue_per_branch)]-revenue_per_branch[np.argmin(revenue_per_branch)]}.")
```

Challenge 2

```python
np.random.seed(33)
data=np.arange(1,101)
np.random.shuffle(data)
data=data.reshape(10,10)
print("Data:\n",data)
```

```python
# Calculate digonal sum.
diag_sum=np.trace(data)
print("Main diagonal sum: ",diag_sum)
```

```python
# Find all positions where values are divisible by both 3 and 5.

# masking
mask=(data%3==0) & (data%5==0)

# indices
rows,cols=np.where(mask)

for r,c in zip(rows,cols):
    print(f"Value: {data[r,c]} , Position: ({r},{c})")
```

```python
# Extract those values and rank then from highest to lowest.
extracted_values=data[mask]
print("Extracted values: ",extracted_values)

rank=np.argsort(extracted_values)[::-1]
print("Ranked: ",extracted_values[rank])
```

```python
# Determine the median of the extracted values.
print("The median of the extracted values: ",np.median(extracted_values))
```

```python
# Find the position of the largest extracted value within the original matrix.
r,c=np.where(extracted_values.max()==data)
print(f"Largest extracted value: {data[r.item(),c.item()]} , Position in original matrix: ({r.item()},{c.item()})")
```

Challenge 3

```python
marks = np.array([
    [45, 78, 92, 35, 88],
    [55, 67, 81, 49, 91],
    [39, 95, 72, 84, 100],
    [61, 58, 77, 89, 42],
    [90, 83, 65, 70, 96]
])
```

```python
# Marks below 40 becomes 40.
# Marks above 95 becomes 95.
polished_marks=np.clip(marks,45,95)
print("Polished marks:\n",polished_marks)
```

```python
# Determine all correccted positions.
rows,cols=np.where(marks!=polished_marks)
print("Corrected Positions are: ")
for r,c in zip(rows,cols):
    print(f"Positions: ({r},{c})     {polished_marks[r,c]}(Currently), {marks[r,c]}(Original)")
```

```python
# Find the topper after correction.
total_marks=polished_marks.sum(axis=0) # columns->students
topper=np.argmax(total_marks)
print(f"Topper after correctection is Student{topper+1} with total marks: {total_marks[np.argmax(total_marks)]}")
```

```python
# Rank all students by total marks.
rank=np.argsort(total_marks)[::-1]
print("Student rank according to total marks: ",rank+1)
```

```python
# Find the top 25%% scores in the dataset.
scores_flat=polished_marks.flatten()
aranged_scores=np.sort(scores_flat)[::-1]
top25=int(0.25*scores_flat.size)
print("Top 25% scores: ",aranged_scores[:top25])
```

```python
# Find all positions containing values between 85 and 95 inclusive.
rows,cols=np.where((85<=polished_marks) & (polished_marks<=95))
print("Positions containg scores between 85 and 95 inclusive are: \n")
for r,c in zip(rows,cols):
    print(f"Position: ({r},{c})     Value: {polished_marks[r,c]}")
```

```python
# Calculate the percentage of scores above the overall mean.
mean_score=polished_marks.mean()
scores_above_avg=np.sum(polished_marks>mean_score)
percentage=(scores_above_avg/polished_marks.size)*100
print(f"Percentage of scores above the overall mean: {percentage:.2f}%")
```

```python
# Identify the most consistent student.
lowest_mark_per_student=polished_marks.min(axis=0)
highest_mark_per_student=polished_marks.max(axis=0)
difference=highest_mark_per_student-lowest_mark_per_student
most_consistent_student=np.argmin(difference)
print("The most consitent student is Student",most_consistent_student+1," with lowest mark variance of ",difference[most_consistent_student])
```
