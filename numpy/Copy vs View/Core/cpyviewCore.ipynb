# An array of empoyees salaries:

```python
import numpy as np 
salaries=np.array([45000,52000,61000,38000,70000,48000],dtype=float)
```

```python
# A colleague slices the salaries to get first 4 Salaries.
salaries4=salaries[:4]
print("Sliced first 4 Salaries: ",salaries4)
```

```python
# Everyone in the slice get 10% raise
salaries4*=1.1
print("Sliced raise salaries: ",salaries4)
```

```python
# Will the original salaries array reflect reflect this change?
print("Orginal Salaries: ",salaries) # Affected
```

```python
# Fix it so the orgiginal is not affected.
salaries=np.array([45000,52000,61000,38000,70000,48000],dtype=float)
salaries4=salaries[:4].copy()
salaries4*=1.1
print("Raised Salaries: ",salaries4)
print("Orginal Salaries: ",salaries)
```

# A 2D array

```python
data = np.arange(1, 13).reshape(3, 4)
```

```python
# Take the transpose of data and assign it to t.
t=np.transpose(data)
print("Orginal data: \n",data)
print("Assigned Transpose data: \n",t)
```

```python
# Modify some element in t and check whether data was affected.
t[0,0]=100
t[1,1]=200
print("Transposed data after modification: \n",t)
print("Original data: \n",data)
```

Both original data is also affected because we assigned t the transpose value of data
 so t referenced to data when we modified.

# Two Variables pointing to the same array

```python
a = np.array([10, 20, 30, 40, 50])
b = a
c = a.copy()
print(np.shares_memory(a,b))
print(np.shares_memory(a,c))
```

# A dataset of 6 products with their stock quantities:

```python
inventory = np.array([
    [101, 340],
    [102,  85],
    [103, 512],
    [104,  20],
    [105, 178],
    [106,  95]
])
```

```python
# Split it into halves
half1,half2=np.vsplit(inventory,2)
print("First half products: \n",half1)
print("Second half products: \n",half2)
```

```python
# From the bottom half, create a view by slicing only the quantity column.
view=half2[:,1]
print("View: \n", view)
```

```python
# Set all quantities below 100 to 0 using Boolean indexing on that view.
view[view<100]=0
print("The view: \n",view)
```

```python
# Check if the original inventory was modified.
print("Inventory data: \n",inventory) # yes
```

# An 1D array of 12 monthly revenue values

```python
revenue = np.array([500, 430, 610, 720, 680, 540, 490, 770, 630, 580, 710, 650])
```

```python
# User ravel to flatten and assign to flat.
flat=revenue.ravel()
print("Raveled Array: ",flat)
```

```python
# Confirm it's a view using .base.
print("Is base?: ",flat.base is None) # false represents it's a view
```

```python
# Use flatten() to crate a flat_copy.
flat_copy=revenue.flatten()
print("Flat Copy: ",flat_copy)
```

```python
# Modify first elements of flat and flat_copy.
flat[0]=999
flat_copy[0]=999
print("Flat: ",flat )
print("Flat Copy: ",flat_copy )
```

```python
# Findal state of revenue
print("Revenue : ",revenue)
print("flat cause change?: ",flat.base is None)
print("flat_copy cause change?: ",flat_copy.base is None)
```

# A student score array

```python
scores = np.array([
    [88, 76, 91],
    [72, 85, 68],
    [95, 90, 87],
    [60, 70, 75]
])
```

```python
# Use fancy indexing to select rows [0,2] and assign to selected.
selected=scores[[0,2],:]
print("Filtered Row(0,2): \n",selected)
```

```python
# Mofify selected[0,0] to 0
selected[0,0]=0
print("Filtered row after modification: \n",selected)
```

```python
# Check if scores was affected using np.shares_memory().
print("Scores affected?: ",np.shares_memory(scores,selected))
```
