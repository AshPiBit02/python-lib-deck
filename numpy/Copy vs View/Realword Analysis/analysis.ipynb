# Preprocessing a dataset for a machine learning pipeline:

```python
import numpy as np 
np.random.seed(0)
dataset = np.random.randint(1, 100, size=(8, 5))
print("Datasets: \n",dataset)
```

```python
# Split into train(6) and test(2).
train,test=np.vsplit(dataset,[6])
print("Train datasets: \n",train)
print("Test datasets: \n",test)
```

```python
# Normalize the train set by subtracting its mean and dividing by its std - do this on a copy so the original dataset is preserved
trainMean=train.mean()
trainStd=train.std()
print("Train dataset mean: ",np.round(trainMean,2))
print("Train dataset standard deviation: ",np.round(trainStd),2)
trainsetCpy=train.copy()
normalizedTrainset=np.round((train-trainMean)/trainStd,2)
print("Original Train Datasets:\n ",train)
print("Normalized Train Datasets:\n ",normalizedTrainset)
```

```python
# Verify the original is unchanged using np.shares_memory()
print("Is changed? : ",np.shares_memory(normalizedTrainset,train)) # as we have changed the train data if they both had referred to same memory, the data would had modified in original set also.
```

# Received sensor readings and need to apply corrections without losing raw data:

```python
readings = np.array([
    [1.02, 0.98, 1.05, 0.97],
    [1.10, 1.03, 0.95, 1.08],
    [0.93, 1.07, 1.02, 0.99],
    [1.05, 0.96, 1.08, 1.01]
])
```

```python
# Create a working copy of readings.
data=readings.copy()
print("Copy readings data: \n",data)
```

```python
# On the copy, clip all values to the range [0.95,1.05]
clipped_data=np.clip(data,0.95,1.05)
print("Clippled data readings: \n",clipped_data)
```

```python
# Reshape the clipped copy into shape(2,8)
reshaped_data=clipped_data.reshape(2,8)
print("Reshaped clipped data: \n",reshaped_data)
```

```python
# Verify the original is completely untouched.
print("Is data copy of original?: ",np.shares_memory(readings,data))
print("Is a copy?: ",data.base is None)
# The shares memory varifies that data has copy of readings. So it is not possible to modify readings dataset by modifying data.
# Also, data is not referencing to readings.
```
